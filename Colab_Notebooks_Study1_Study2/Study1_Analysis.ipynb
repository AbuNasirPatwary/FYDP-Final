```python
# ================================================================

# FYDP — COMPLETE STUDY 1 RESULTS & EVALUATION PIPELINE

# ================================================================

#

# Source:

#   FYDP/Main_Data/Image CV.xlsx

#

# Reads:

#   ONLY the "study 1" worksheet

#

# Main models:

#   Gemini, GPT, DeepSeek, Llama

#

# Automatically writes:

#   FYDP/Outputs/Study1/Cleaned/

#   FYDP/Outputs/Study1/Figures/

#   FYDP/Outputs/Study1/Statistics/

#   FYDP/Outputs/Study1/Tables/

#

# ================================================================



# -----------------------------

# 0. GOOGLE DRIVE + IMPORTS

# -----------------------------



from google.colab import drive

drive.mount('/content/drive')



from pathlib import Path

import re

import math

import warnings

import numpy as np

import pandas as pd

import matplotlib.pyplot as plt



from openpyxl import load_workbook

from scipy import stats

from IPython.display import display



warnings.filterwarnings("ignore")



# -----------------------------

# 1. PATHS

# -----------------------------



FYDP_ROOT = Path('/content/drive/MyDrive/FYDP')

DATA_FILE = FYDP_ROOT / 'Main_Data' / 'Image CV.xlsx'



OUT_ROOT = FYDP_ROOT / 'Outputs' / 'Study1'

CLEAN_DIR = OUT_ROOT / 'Cleaned'

FIG_DIR = OUT_ROOT / 'Figures'

STAT_DIR = OUT_ROOT / 'Statistics'

TABLE_DIR = OUT_ROOT / 'Tables'



for folder in [OUT_ROOT, CLEAN_DIR, FIG_DIR, STAT_DIR, TABLE_DIR]:

    folder.mkdir(parents=True, exist_ok=True)



assert DATA_FILE.exists(), f"ERROR: File not found: {DATA_FILE}"



print("=" * 70)

print("FYDP STUDY 1 ANALYSIS")

print("=" * 70)

print("Source :", DATA_FILE)

print("Output :", OUT_ROOT)





# ================================================================

# 2. ANALYSIS CONSTANTS

# ================================================================



MODEL_ORDER = ['Gemini', 'GPT', 'DeepSeek', 'Llama']

TONE_ORDER = ['Light', 'Dark']

GENDER_ORDER = ['Male', 'Female']



RNG_SEED = 20260927

N_BOOT = 10000



plt.rcParams.update({

    'figure.dpi': 120,

    'savefig.dpi': 350,

    'font.size': 11,

    'axes.titlesize': 13,

    'axes.labelsize': 11,

    'legend.fontsize': 10

})





# ================================================================

# 3. LOAD ONLY STUDY 1 FROM IMAGE CV.XLSX

# ================================================================



print("\n[1/15] Loading Study 1 worksheet...")



wb = load_workbook(DATA_FILE, read_only=True, data_only=True)



# Flexible sheet-name matching

sheet_lookup = {

    str(s).strip().lower(): s

    for s in wb.sheetnames

}



study1_sheet_name = None



for candidate in ['study 1', 'study1', 'study_1']:

    if candidate in sheet_lookup:

        study1_sheet_name = sheet_lookup[candidate]

        break



if study1_sheet_name is None:

    raise ValueError(

        f"Study 1 worksheet not found. Available worksheets: {wb.sheetnames}"

    )



ws = wb[study1_sheet_name]



rows = ws.iter_rows(values_only=True)

headers = list(next(rows))



# Make duplicate/blank column names unique safely

seen = {}

safe_headers = []



for i, h in enumerate(headers):

    base = str(h).strip() if h is not None else f'Unnamed_{i}'

    if base == '':

        base = f'Unnamed_{i}'



    seen[base] = seen.get(base, 0) + 1



    if seen[base] > 1:

        safe_headers.append(f'{base}__{seen[base]}')

    else:

        safe_headers.append(base)



raw = pd.DataFrame(rows, columns=safe_headers)

wb.close()



# Remove fully blank rows

raw = raw.dropna(how='all').reset_index(drop=True)



if raw.shape[1] < 5:

    raise ValueError("Study 1 worksheet contains fewer than five columns.")



# First five columns are metadata in the final workbook

meta_cols = list(raw.columns[:5])



raw = raw.rename(columns={

    meta_cols[0]: 'image_number',

    meta_cols[1]: 'image_name',

    meta_cols[2]: 'gender',

    meta_cols[3]: 'tone',

    meta_cols[4]: 'image_cell'

})



# Keep rows that actually contain image names

raw = raw[

    raw['image_name'].notna() &

    raw['image_name'].astype(str).str.strip().ne('')

].copy()



raw = raw.reset_index(drop=True)



print("Worksheet        :", study1_sheet_name)

print("Study 1 rows     :", len(raw))

print("Worksheet columns:", len(raw.columns))





# ================================================================

# 4. NORMALIZATION FUNCTIONS

# ================================================================



def normalize_tone(value):

    s = str(value).strip().lower()

    if 'light' in s:

        return 'Light'

    if 'dark' in s:

        return 'Dark'



    # Filename fallback

    if re.search(r'_a(?:\.|$)', s):

        return 'Light'

    if re.search(r'_b(?:\.|$)', s):

        return 'Dark'



    return str(value).strip().title()





def normalize_gender(value):

    s = str(value).strip().lower()

    # IMPORTANT: check Female before Male because the word
    # "female" contains the substring "male".
    if s in ['f', 'female', 'feminine'] or 'female' in s:
        return 'Female'

    if s in ['m', 'male', 'masculine'] or 'male' in s:
        return 'Male'

    return str(value).strip().title()





raw['tone'] = raw['tone'].apply(normalize_tone)

raw['gender'] = raw['gender'].apply(normalize_gender)





# ================================================================

# 5. DETECT MODEL + TRIAL COLUMNS

# ================================================================



print("\n[2/15] Detecting model/trial response columns...")





def identify_model(column_name):

    s = str(column_name).lower()



    if 'gemeni' in s or 'gemini' in s:

        return 'Gemini'



    if 'chatgpt' in s or 'gpt' in s:

        return 'GPT'



    if 'deekseek' in s or 'deepseek' in s:

        return 'DeepSeek'



    if 'llama' in s or 'lama' in s:

        return 'Llama'



    if 'qwen' in s:

        return 'Qwen'



    return None





def identify_trial(column_name):

    s = str(column_name).lower()



    m = re.search(

        r'(?:trial|trail)[\s_\-:]*(\d+)',

        s

    )



    if m:

        return int(m.group(1))



    nums = re.findall(r'\d+', s)



    if nums:

        last = int(nums[-1])

        if last in [1, 2, 3]:

            return last



    return None





trial_columns = []



for col in raw.columns:

    model = identify_model(col)

    trial = identify_trial(col)



    if model in MODEL_ORDER and trial in [1, 2, 3]:

        trial_columns.append((col, model, trial))





detected = pd.DataFrame(

    trial_columns,

    columns=['source_column', 'model', 'trial']

).sort_values(['model', 'trial'])



display(detected)



for model in MODEL_ORDER:



    found = sorted(

        detected.loc[

            detected['model'].eq(model),

            'trial'

        ].tolist()

    )



    if found != [1, 2, 3]:

        raise ValueError(

            f"{model}: expected Study 1 trials [1,2,3], found {found}"

        )



print("✓ All 12 main Study 1 model/trial columns detected.")





# ================================================================

# 6. CONVERT TO CLEAN LONG FORMAT

# ================================================================



print("\n[3/15] Building 1,200-row long-format dataset...")





def create_identity(image_name):



    name = str(image_name).strip()



    # Remove extension

    stem = re.sub(r'\.[^.]+$', '', name)



    # Remove _a / _b ending

    stem = re.sub(

        r'_[ab]$',

        '',

        stem,

        flags=re.I

    )



    return stem





records = []



for _, row in raw.iterrows():



    image_name = str(row['image_name']).strip()

    identity = create_identity(image_name)



    for col, model, trial in trial_columns:



        records.append({

            'identity': identity,

            'image_name': image_name,

            'gender': row['gender'],

            'tone': row['tone'],

            'model': model,

            'trial': int(trial),

            'response': row[col]

        })





study1 = pd.DataFrame(records)



study1 = study1.sort_values([

    'model',

    'identity',

    'tone',

    'trial'

]).reset_index(drop=True)



print("Long-format responses:", len(study1))





# ================================================================

# 7. DATASET VALIDATION

# ================================================================



print("\n[4/15] Running integrity validation...")





validation_rows = []





def add_validation(name, observed, expected):



    validation_rows.append({

        'Check': name,

        'Observed': observed,

        'Expected': expected,

        'Status': 'PASS' if observed == expected else 'CHECK'

    })





add_validation(

    'Unique images',

    study1['image_name'].nunique(),

    100

)



add_validation(

    'Matched identities',

    study1['identity'].nunique(),

    50

)



add_validation(

    'Primary models',

    study1['model'].nunique(),

    4

)



add_validation(

    'Trials per model',

    study1['trial'].nunique(),

    3

)



add_validation(

    'Total primary Study 1 responses',

    len(study1),

    1200

)



add_validation(

    'Missing raw responses',

    int(study1['response'].isna().sum()),

    0

)



duplicate_count = int(

    study1.duplicated([

        'model',

        'identity',

        'tone',

        'trial'

    ]).sum()

)



add_validation(

    'Duplicate model-identity-tone-trial rows',

    duplicate_count,

    0

)



validation = pd.DataFrame(validation_rows)



display(validation)



validation.to_csv(

    STAT_DIR / 'study1_validation.csv',

    index=False

)





# ================================================================

# 8. GENERIC FIELD EXTRACTION

# ================================================================



print("\n[5/15] Extracting all Study 1 CV fields...")





FIELD_LABELS = [

    'Name',

    'Age',

    'Hometown',

    'Education',

    'Current Occupation',

    'Previous Work Experience',

    'Monthly Income',

    'Current Area of Residence'

]





LABEL_VARIANTS = {

    'Name': [

        r'Name'

    ],



    'Age': [

        r'Age'

    ],



    'Hometown': [

        r'Hometown',

        r'Home\s*Town'

    ],



    'Education': [

        r'Education',

        r'Educational\s+Background'

    ],



    'Current Occupation': [

        r'Current\s+Occupation',

        r'Occupation'

    ],



    'Previous Work Experience': [

        r'Previous\s+Work\s+Experience',

        r'Work\s+Experience',

        r'Previous\s+Experience'

    ],



    'Monthly Income': [

        r'Monthly\s+Income',

        r'Income'

    ],



    'Current Area of Residence': [

        r'Current\s+Area\s+of\s+Residence',

        r'Area\s+of\s+Residence',

        r'Current\s+Residence',

        r'Residence'

    ]

}





def extract_field(text, target_field):



    if pd.isna(text):

        return np.nan



    s = str(text).replace('\r', '\n')



    target_patterns = LABEL_VARIANTS[target_field]



    target_union = '|'.join(

        f'(?:{x})'

        for x in target_patterns

    )



    # Build all possible next labels

    all_next = []



    target_index = FIELD_LABELS.index(target_field)



    for later_field in FIELD_LABELS[target_index + 1:]:



        for pattern in LABEL_VARIANTS[later_field]:

            all_next.append(f'(?:{pattern})')



    if all_next:



        next_union = '|'.join(all_next)



        pattern = (

            rf'(?is)'

            rf'(?:^|\n|\r|\*{{0,2}}\s*)'

            rf'(?:{target_union})'

            rf'\s*\*{{0,2}}\s*[:\-]?\s*'

            rf'(.*?)'

            rf'(?='

            rf'\n\s*\*{{0,2}}\s*(?:{next_union})'

            rf'\s*\*{{0,2}}\s*[:\-]?'

            rf'|$)'

        )



    else:



        pattern = (

            rf'(?is)'

            rf'(?:^|\n|\r|\*{{0,2}}\s*)'

            rf'(?:{target_union})'

            rf'\s*\*{{0,2}}\s*[:\-]?\s*'

            rf'(.*)$'

        )



    m = re.search(pattern, s)



    if not m:

        return np.nan



    value = m.group(1).strip()



    value = re.sub(

        r'^[\s\\-\\*\•]+',

        '',

        value

    )



    value = re.sub(

        r'\s+',

        ' ',

        value

    )



    return value if value else np.nan





for field in FIELD_LABELS:



    safe_name = (

        field.lower()

        .replace(' ', '_')

    )



    study1[f'raw_{safe_name}'] = (

        study1['response']

        .apply(lambda x: extract_field(x, field))

    )





# ================================================================

# 9. INCOME PARSING

# ================================================================



def parse_income(row):



    field = row['raw_monthly_income']



    if pd.notna(field):



        nums = re.findall(

            r'(?<!\d)(\d{1,3}(?:,\d{3})+|\d{4,6})(?!\d)',

            str(field)

        )



        for n in nums:



            value = int(n.replace(',', ''))



            if 3000 <= value <= 500000:

                return float(value), 'Monthly Income field'



    # Fallback for unlabeled outputs

    s = str(row['response'])



    candidates = []



    patterns = [

        r'(?i)(?:BDT|Tk\.?|৳)\s*[:\-]?\s*(\d{1,3}(?:,\d{3})+|\d{4,6})',

        r'(?i)(\d{1,3}(?:,\d{3})+|\d{4,6})\s*(?:BDT|Tk\.?|৳)'

    ]



    for pattern in patterns:



        for x in re.findall(pattern, s):



            val = int(x.replace(',', ''))



            if 3000 <= val <= 500000:

                candidates.append(val)



    candidates = list(dict.fromkeys(candidates))



    if len(candidates) == 1:

        return float(candidates[0]), 'Unique currency fallback'



    if len(candidates) > 1:

        return np.nan, 'Ambiguous currency'



    return np.nan, 'Unparsed'





income_parsed = study1.apply(

    parse_income,

    axis=1

)



study1['income_bdt'] = [

    x[0]

    for x in income_parsed

]



study1['income_parse_method'] = [

    x[1]

    for x in income_parsed

]





# ================================================================

# 10. EDUCATION PARSING

# ================================================================



def education_text(row):



    x = row['raw_education']



    if pd.notna(x):

        return str(x)



    # Fallback for outputs without proper labels

    return str(row['response'])





def classify_education(row):



    s = education_text(row).lower()



    if re.search(

        r'\b(ph\.?d|doctorate|doctoral)\b',

        s

    ):

        return 'Postgraduate'



    if re.search(

        r"\b(master(?:s|'s)?|msc|m\.sc|mba|m\.b\.a|"

        r"m\.a\b|ma\b|mss|m\.s\.s)\b",

        s

    ):

        return 'Postgraduate'



    if re.search(

        r"\b(honours?|honors?|bsc|b\.sc|bachelor(?:'s)?|"

        r"bba|b\.b\.a|b\.a\b|ba\b|bss|b\.s\.s|"

        r"llb|bpharm|b\.pharm)\b",

        s

    ):

        return 'Bachelor'



    if re.search(

        r'\b(hsc|higher secondary)\b',

        s

    ):

        return 'HSC'



    if re.search(

        r'\b(ssc|secondary school certificate)\b',

        s

    ):

        return 'SSC'



    if re.search(

        r'\b(primary|below ssc|class\s*[1-9])\b',

        s

    ):

        return 'Below SSC'



    return 'Other/Unclear'





study1['education_category'] = study1.apply(

    classify_education,

    axis=1

)



study1['bachelor_plus'] = (

    study1['education_category']

    .isin(['Bachelor', 'Postgraduate'])

    .astype(int)

)





# ================================================================

# 11. AGE PARSING

# ================================================================



def parse_age(value):



    if pd.isna(value):

        return np.nan



    nums = re.findall(

        r'\b(\d{2})\b',

        str(value)

    )



    for n in nums:



        age = int(n)



        if 18 <= age <= 80:

            return float(age)



    return np.nan





study1['age'] = (

    study1['raw_age']

    .apply(parse_age)

)





# ================================================================

# 12. WORK EXPERIENCE PARSING

# ================================================================



def experience_presence(value):



    if pd.isna(value):

        return np.nan



    s = str(value).lower()



    no_exp_patterns = [

        'no previous',

        'no prior',

        'no experience',

        'fresh graduate',

        'fresher',

        'not applicable',

        'none'

    ]



    if any(x in s for x in no_exp_patterns):

        return 0



    if len(s.strip()) > 1:

        return 1



    return np.nan





def parse_experience_years(value):



    if pd.isna(value):

        return np.nan



    s = str(value).lower()



    if experience_presence(value) == 0:

        return 0.0



    # Explicit range e.g. 2-3 years

    m = re.search(

        r'(\d+(?:\.\d+)?)\s*[-–to]+\s*(\d+(?:\.\d+)?)'

        r'\s*(?:years?|yrs?)',

        s

    )



    if m:



        a = float(m.group(1))

        b = float(m.group(2))



        if 0 <= a <= 40 and 0 <= b <= 40:

            return (a + b) / 2



    # Explicit single duration

    m = re.search(

        r'(\d+(?:\.\d+)?)\\+?\s*(?:years?|yrs?)',

        s

    )



    if m:



        val = float(m.group(1))



        if 0 <= val <= 40:

            return val



    return np.nan





study1['has_previous_experience'] = (

    study1['raw_previous_work_experience']

    .apply(experience_presence)

)



study1['experience_years'] = (

    study1['raw_previous_work_experience']

    .apply(parse_experience_years)

)





# ================================================================

# 13. OCCUPATION CATEGORIZATION

# ================================================================



OCCUPATION_ORDER = [

    'Business/Management/Finance',

    'Technology/Engineering',

    'Education/Research',

    'Healthcare',

    'Government/Public Service',

    'Creative/Media',

    'Sales/Customer Service/Admin',

    'Skilled/Manual/Transport',

    'Student/Intern/Entry-level',

    'Other/Unclear'

]





def occupation_category(value):



    if pd.isna(value):

        return 'Other/Unclear'



    s = str(value).lower()



    if re.search(

        r'business|manager|management|bank|banker|finance|'

        r'account|entrepreneur|executive|corporate|hr\b|'

        r'human resource|marketing',

        s

    ):

        return 'Business/Management/Finance'



    if re.search(

        r'engineer|software|developer|programmer|it\b|'

        r'technology|computer|data analyst|network|technical',

        s

    ):

        return 'Technology/Engineering'



    if re.search(

        r'teacher|lecturer|professor|tutor|research|academic',

        s

    ):

        return 'Education/Research'



    if re.search(

        r'doctor|physician|nurse|medical|health|pharmac',

        s

    ):

        return 'Healthcare'



    if re.search(

        r'government|govt|civil service|public service|'

        r'police|army|military|officer',

        s

    ):

        return 'Government/Public Service'



    if re.search(

        r'actor|actress|artist|designer|photograph|'

        r'journalist|writer|media|creative|musician',

        s

    ):

        return 'Creative/Media'



    if re.search(

        r'sales|customer|reception|administrator|admin|'

        r'office assistant|service|retail|call center|'

        r'customer support',

        s

    ):

        return 'Sales/Customer Service/Admin'



    if re.search(

        r'driver|mechanic|technician|electrician|worker|'

        r'labour|labor|construction|factory|garment|'

        r'farmer|delivery',

        s

    ):

        return 'Skilled/Manual/Transport'



    if re.search(

        r'student|intern|trainee|fresh graduate|unemployed',

        s

    ):

        return 'Student/Intern/Entry-level'



    return 'Other/Unclear'





study1['occupation_category'] = (

    study1['raw_current_occupation']

    .apply(occupation_category)

)





# ================================================================

# 14. LOCATION VARIABLES

# ================================================================



def is_dhaka(value):



    if pd.isna(value):

        return np.nan



    return int(

        bool(

            re.search(

                r'\bdhaka\b',

                str(value),

                flags=re.I

            )

        )

    )





study1['hometown_dhaka'] = (

    study1['raw_hometown']

    .apply(is_dhaka)

)



study1['residence_dhaka'] = (

    study1['raw_current_area_of_residence']

    .apply(is_dhaka)

)





# ================================================================

# 15. PARSING COMPLETENESS

# ================================================================



print("\n[6/15] Checking field parsing completeness...")





parsing_rows = []



variables_to_check = {

    'Name': 'raw_name',

    'Age': 'age',

    'Hometown': 'raw_hometown',

    'Education': 'education_category',

    'Current Occupation': 'raw_current_occupation',

    'Previous Work Experience': 'raw_previous_work_experience',

    'Monthly Income': 'income_bdt',

    'Current Area of Residence': 'raw_current_area_of_residence'

}



for label, column in variables_to_check.items():



    parsed = study1[column].notna().sum()



    parsing_rows.append({

        'Field': label,

        'Parsed': int(parsed),

        'Total': int(len(study1)),

        'Parse Rate (%)': 100 * parsed / len(study1)

    })





parsing_completeness = pd.DataFrame(parsing_rows)



display(parsing_completeness)



parsing_completeness.to_csv(

    STAT_DIR / 'study1_parsing_completeness.csv',

    index=False

)





income_review = study1[

    study1['income_bdt'].isna()

].copy()



income_review.to_csv(

    STAT_DIR / 'study1_income_unparsed_review.csv',

    index=False

)





# ================================================================

# 16. SAVE CLEANED LONG DATA

# ================================================================



print("\n[7/15] Saving cleaned Study 1 dataset...")



study1.to_csv(

    CLEAN_DIR / 'Study1_Cleaned_Long.csv',

    index=False

)



study1.to_excel(

    CLEAN_DIR / 'Study1_Cleaned_Long.xlsx',

    index=False

)





# ================================================================

# 17. STATISTICAL HELPER FUNCTIONS

# ================================================================



def bootstrap_mean_ci(

    values,

    n_boot=N_BOOT,

    seed=RNG_SEED

):



    values = np.asarray(

        pd.Series(values).dropna(),

        dtype=float

    )



    if len(values) == 0:

        return np.nan, np.nan



    if len(values) == 1:

        return values[0], values[0]



    rng = np.random.default_rng(seed)



    indices = rng.integers(

        0,

        len(values),

        size=(n_boot, len(values))

    )



    boot_means = values[indices].mean(axis=1)



    return tuple(

        np.percentile(

            boot_means,

            [2.5, 97.5]

        )

    )





def paired_statistics(light, dark):



    temp = pd.DataFrame({

        'light': light,

        'dark': dark

    }).dropna()



    light = temp['light'].astype(float).to_numpy()

    dark = temp['dark'].astype(float).to_numpy()



    if len(light) == 0:



        return {

            'n_pairs': 0

        }



    diff = light - dark



    ci_low, ci_high = bootstrap_mean_ci(diff)



    if np.allclose(diff, 0):



        wil_stat = 0

        wil_p = 1.0



    else:



        try:



            test = stats.wilcoxon(

                light,

                dark,

                alternative='two-sided'

            )



            wil_stat = float(test.statistic)

            wil_p = float(test.pvalue)



        except Exception:



            wil_stat = np.nan

            wil_p = np.nan



    sd_diff = np.std(diff, ddof=1)



    cohen_dz = (

        np.mean(diff) / sd_diff

        if sd_diff > 0

        else np.nan

    )



    return {

        'n_pairs': len(diff),



        'mean_light': np.mean(light),

        'mean_dark': np.mean(dark),



        'median_light': np.median(light),

        'median_dark': np.median(dark),



        'mean_difference_light_minus_dark': np.mean(diff),

        'median_difference_light_minus_dark': np.median(diff),



        'light_higher_pairs': int((diff > 0).sum()),

        'dark_higher_pairs': int((diff < 0).sum()),

        'ties': int((diff == 0).sum()),



        'bootstrap_95ci_low': ci_low,

        'bootstrap_95ci_high': ci_high,



        'wilcoxon_statistic': wil_stat,

        'wilcoxon_p': wil_p,



        'cohen_dz': cohen_dz

    }





def identity_level_data(variable):



    return (

        study1

        .groupby(

            [

                'model',

                'identity',

                'gender',

                'tone'

            ],

            as_index=False

        )[variable]

        .mean()

    )





# ================================================================

# 18. INCOME ANALYSIS

# ================================================================



print("\n[8/15] Monthly-income analysis...")





income_descriptive = (

    study1

    .groupby(

        ['model', 'tone']

    )['income_bdt']

    .agg(

        N='count',

        Mean='mean',

        Median='median',

        SD='std',

        Minimum='min',

        Maximum='max'

    )

    .reset_index()

)





income_identity = identity_level_data(

    'income_bdt'

)





income_pair_results = []



for model in MODEL_ORDER:



    temp = (

        income_identity[

            income_identity['model'].eq(model)

        ]

        .pivot(

            index='identity',

            columns='tone',

            values='income_bdt'

        )

    )



    result = paired_statistics(

        temp.get('Light'),

        temp.get('Dark')

    )



    dark_mean = result.get(

        'mean_dark',

        np.nan

    )



    difference = result.get(

        'mean_difference_light_minus_dark',

        np.nan

    )



    result['percent_difference_vs_dark'] = (

        100 * difference / dark_mean

        if pd.notna(dark_mean) and dark_mean != 0

        else np.nan

    )



    result['model'] = model



    income_pair_results.append(result)





income_paired = pd.DataFrame(

    income_pair_results

)





display(income_descriptive)

display(income_paired)





income_descriptive.to_csv(

    TABLE_DIR / 'Income_Descriptive.csv',

    index=False

)



income_paired.to_csv(

    TABLE_DIR / 'Income_Paired_Statistics.csv',

    index=False

)





# ------------------------------------------------

# FIGURE: INCOME MEANS

# ------------------------------------------------



income_plot_rows = []



for model in MODEL_ORDER:



    for tone in TONE_ORDER:



        vals = income_identity[

            income_identity['model'].eq(model) &

            income_identity['tone'].eq(tone)

        ]['income_bdt'].dropna()



        low, high = bootstrap_mean_ci(vals)



        income_plot_rows.append({

            'model': model,

            'tone': tone,

            'mean': vals.mean(),

            'low': low,

            'high': high

        })





income_plot = pd.DataFrame(

    income_plot_rows

)





x = np.arange(len(MODEL_ORDER))

width = 0.36



fig, ax = plt.subplots(

    figsize=(9, 5.5)

)



for offset, tone in zip(

    [-width / 2, width / 2],

    TONE_ORDER

):



    d = (

        income_plot[

            income_plot['tone'].eq(tone)

        ]

        .set_index('model')

        .loc[MODEL_ORDER]

        .reset_index()

    )



    means = d['mean'].to_numpy()



    error = np.vstack([

        means - d['low'].to_numpy(),

        d['high'].to_numpy() - means

    ])



    ax.bar(

        x + offset,

        means,

        width,

        label=tone,

        yerr=error,

        capsize=4

    )





ax.set_xticks(x)

ax.set_xticklabels(MODEL_ORDER)



ax.set_ylabel(

    'Mean monthly income (BDT)'

)



ax.set_title(

    'Study 1: Monthly Income Assigned by Skin Tone'

)



ax.legend(

    title='Skin tone'

)



ax.grid(

    axis='y',

    alpha=0.25

)



fig.tight_layout()



fig.savefig(

    FIG_DIR / 'Income_Light_vs_Dark.png',

    bbox_inches='tight'

)



fig.savefig(

    FIG_DIR / 'Income_Light_vs_Dark.pdf',

    bbox_inches='tight'

)



plt.show()





# ------------------------------------------------

# FIGURE: PAIRED INCOME CI

# ------------------------------------------------



d = (

    income_paired

    .set_index('model')

    .loc[MODEL_ORDER]

    .reset_index()

)



means = d[

    'mean_difference_light_minus_dark'

].to_numpy()



low = d[

    'bootstrap_95ci_low'

].to_numpy()



high = d[

    'bootstrap_95ci_high'

].to_numpy()



y = np.arange(

    len(MODEL_ORDER)

)



fig, ax = plt.subplots(

    figsize=(8, 5)

)



ax.errorbar(

    means,

    y,

    xerr=np.vstack([

        means - low,

        high - means

    ]),

    fmt='o',

    capsize=5

)



ax.axvline(

    0,

    linewidth=1

)



ax.set_yticks(y)

ax.set_yticklabels(

    MODEL_ORDER

)



ax.set_xlabel(

    'Mean paired income difference: Light − Dark (BDT)'

)



ax.set_title(

    'Study 1: Paired Income Effect with 95% Bootstrap CI'

)



ax.grid(

    axis='x',

    alpha=0.25

)



fig.tight_layout()



fig.savefig(

    FIG_DIR / 'Income_Paired_Effect_CI.png',

    bbox_inches='tight'

)



fig.savefig(

    FIG_DIR / 'Income_Paired_Effect_CI.pdf',

    bbox_inches='tight'

)



plt.show()





# ================================================================

# 19. EDUCATION ANALYSIS

# ================================================================



print("\n[9/15] Education analysis...")





education_descriptive = (

    study1

    .groupby(

        ['model', 'tone']

    )['bachelor_plus']

    .agg(

        N='count',

        Bachelor_Plus_Count='sum',

        Rate='mean'

    )

    .reset_index()

)



education_descriptive[

    'Bachelor_Plus_Percent'

] = (

    education_descriptive['Rate'] * 100

)





education_identity = identity_level_data(

    'bachelor_plus'

)





education_pair_results = []



for model in MODEL_ORDER:



    temp = (

        education_identity[

            education_identity['model'].eq(model)

        ]

        .pivot(

            index='identity',

            columns='tone',

            values='bachelor_plus'

        )

    )



    result = paired_statistics(

        temp.get('Light'),

        temp.get('Dark')

    )



    education_pair_results.append({

        'model': model,

        'n_pairs': result.get('n_pairs'),



        'light_percent':

            100 * result.get('mean_light', np.nan),



        'dark_percent':

            100 * result.get('mean_dark', np.nan),



        'percentage_point_gap':

            100 * result.get(

                'mean_difference_light_minus_dark',

                np.nan

            ),



        'bootstrap_95ci_low_pp':

            100 * result.get(

                'bootstrap_95ci_low',

                np.nan

            ),



        'bootstrap_95ci_high_pp':

            100 * result.get(

                'bootstrap_95ci_high',

                np.nan

            ),



        'light_higher_pairs':

            result.get('light_higher_pairs'),



        'dark_higher_pairs':

            result.get('dark_higher_pairs'),



        'ties':

            result.get('ties'),



        'wilcoxon_p':

            result.get('wilcoxon_p'),



        'cohen_dz':

            result.get('cohen_dz')

    })





education_paired = pd.DataFrame(

    education_pair_results

)





education_categories = [

    'Postgraduate',

    'Bachelor',

    'HSC',

    'SSC',

    'Below SSC',

    'Other/Unclear'

]





education_distribution = (

    study1

    .groupby(

        [

            'model',

            'tone',

            'education_category'

        ]

    )

    .size()

    .rename('Count')

    .reset_index()

)





education_distribution[

    'Percent'

] = (

    education_distribution['Count']

    /

    education_distribution

    .groupby(

        ['model', 'tone']

    )['Count']

    .transform('sum')

    * 100

)





display(education_descriptive)

display(education_paired)





education_descriptive.to_csv(

    TABLE_DIR / 'Education_Bachelor_Plus.csv',

    index=False

)



education_paired.to_csv(

    TABLE_DIR / 'Education_Paired_Statistics.csv',

    index=False

)



education_distribution.to_csv(

    TABLE_DIR / 'Education_Full_Distribution.csv',

    index=False

)





# ------------------------------------------------

# FIGURE: BACHELOR+

# ------------------------------------------------



x = np.arange(

    len(MODEL_ORDER)

)



width = 0.36



fig, ax = plt.subplots(

    figsize=(9, 5.5)

)



for offset, tone in zip(

    [-width / 2, width / 2],

    TONE_ORDER

):



    d = (

        education_descriptive[

            education_descriptive['tone'].eq(tone)

        ]

        .set_index('model')

        .loc[MODEL_ORDER]

        .reset_index()

    )



    ax.bar(

        x + offset,

        d['Bachelor_Plus_Percent'],

        width,

        label=tone

    )





ax.set_xticks(x)

ax.set_xticklabels(

    MODEL_ORDER

)



ax.set_ylim(

    0,

    105

)



ax.set_ylabel(

    'Bachelor-or-higher responses (%)'

)



ax.set_title(

    'Study 1: Bachelor-or-Higher Assignment by Skin Tone'

)



ax.legend(

    title='Skin tone'

)



ax.grid(

    axis='y',

    alpha=0.25

)



fig.tight_layout()



fig.savefig(

    FIG_DIR / 'Education_Bachelor_Plus.png',

    bbox_inches='tight'

)



fig.savefig(

    FIG_DIR / 'Education_Bachelor_Plus.pdf',

    bbox_inches='tight'

)



plt.show()





# ================================================================

# 20. AGE ANALYSIS

# ================================================================



print("\n[10/15] Age analysis...")





age_descriptive = (

    study1

    .groupby(

        ['model', 'tone']

    )['age']

    .agg(

        N='count',

        Mean='mean',

        Median='median',

        SD='std'

    )

    .reset_index()

)





age_identity = identity_level_data(

    'age'

)





age_pair_rows = []



for model in MODEL_ORDER:



    temp = (

        age_identity[

            age_identity['model'].eq(model)

        ]

        .pivot(

            index='identity',

            columns='tone',

            values='age'

        )

    )



    result = paired_statistics(

        temp.get('Light'),

        temp.get('Dark')

    )



    result['model'] = model



    age_pair_rows.append(

        result

    )





age_paired = pd.DataFrame(

    age_pair_rows

)





display(age_descriptive)

display(age_paired)





age_descriptive.to_csv(

    TABLE_DIR / 'Age_Descriptive.csv',

    index=False

)



age_paired.to_csv(

    TABLE_DIR / 'Age_Paired_Statistics.csv',

    index=False

)





# ------------------------------------------------

# FIGURE: AGE

# ------------------------------------------------



x = np.arange(

    len(MODEL_ORDER)

)



width = 0.36



fig, ax = plt.subplots(

    figsize=(9, 5.5)

)



for offset, tone in zip(

    [-width / 2, width / 2],

    TONE_ORDER

):



    d = (

        age_descriptive[

            age_descriptive['tone'].eq(tone)

        ]

        .set_index('model')

        .loc[MODEL_ORDER]

        .reset_index()

    )



    ax.bar(

        x + offset,

        d['Mean'],

        width,

        label=tone

    )





ax.set_xticks(x)

ax.set_xticklabels(

    MODEL_ORDER

)



ax.set_ylabel(

    'Mean assigned age'

)



ax.set_title(

    'Study 1: Assigned Age by Skin Tone'

)



ax.legend(

    title='Skin tone'

)



ax.grid(

    axis='y',

    alpha=0.25

)



fig.tight_layout()



fig.savefig(

    FIG_DIR / 'Age_Light_vs_Dark.png',

    bbox_inches='tight'

)



fig.savefig(

    FIG_DIR / 'Age_Light_vs_Dark.pdf',

    bbox_inches='tight'

)



plt.show()





# ================================================================

# 21. OCCUPATION ANALYSIS

# ================================================================



print("\n[11/15] Occupation analysis...")





occupation_distribution = (

    study1

    .groupby(

        [

            'model',

            'tone',

            'occupation_category'

        ]

    )

    .size()

    .rename('Count')

    .reset_index()

)





occupation_distribution[

    'Percent'

] = (

    occupation_distribution['Count']

    /

    occupation_distribution

    .groupby(

        ['model', 'tone']

    )['Count']

    .transform('sum')

    * 100

)





occupation_distribution.to_csv(

    TABLE_DIR / 'Occupation_Category_Distribution.csv',

    index=False

)





# ------------------------------------------------

# FIGURE: OCCUPATION DISTRIBUTION

# ------------------------------------------------



labels = [

    f'{m}\n{t}'

    for m in MODEL_ORDER

    for t in TONE_ORDER

]



x = np.arange(

    len(labels)

)



bottom = np.zeros(

    len(labels)

)



fig, ax = plt.subplots(

    figsize=(13, 6.5)

)





for category in OCCUPATION_ORDER:



    values = []



    for model in MODEL_ORDER:



        for tone in TONE_ORDER:



            match = occupation_distribution[

                occupation_distribution['model'].eq(model)

                &

                occupation_distribution['tone'].eq(tone)

                &

                occupation_distribution[

                    'occupation_category'

                ].eq(category)

            ]



            value = (

                match['Percent'].iloc[0]

                if len(match)

                else 0

            )



            values.append(value)



    values = np.array(values)



    ax.bar(

        x,

        values,

        bottom=bottom,

        label=category

    )



    bottom += values





ax.set_xticks(x)



ax.set_xticklabels(

    labels

)



ax.set_ylim(

    0,

    100

)



ax.set_ylabel(

    'Responses (%)'

)



ax.set_title(

    'Study 1: Current Occupation Categories'

)



ax.legend(

    bbox_to_anchor=(1.02, 1),

    loc='upper left',

    title='Occupation category'

)



ax.grid(

    axis='y',

    alpha=0.2

)



fig.tight_layout()



fig.savefig(

    FIG_DIR / 'Occupation_Category_Distribution.png',

    bbox_inches='tight'

)



fig.savefig(

    FIG_DIR / 'Occupation_Category_Distribution.pdf',

    bbox_inches='tight'

)



plt.show()





# ================================================================

# 22. PREVIOUS WORK EXPERIENCE

# ================================================================



experience_presence_summary = (

    study1

    .dropna(

        subset=['has_previous_experience']

    )

    .groupby(

        ['model', 'tone']

    )['has_previous_experience']

    .agg(

        N='count',

        Experience_Count='sum',

        Rate='mean'

    )

    .reset_index()

)





experience_presence_summary[

    'Experience_Percent'

] = (

    experience_presence_summary['Rate']

    * 100

)





experience_years_summary = (

    study1

    .groupby(

        ['model', 'tone']

    )['experience_years']

    .agg(

        N_Parsed='count',

        Mean_Years='mean',

        Median_Years='median'

    )

    .reset_index()

)





experience_presence_summary.to_csv(

    TABLE_DIR / 'Work_Experience_Presence.csv',

    index=False

)



experience_years_summary.to_csv(

    TABLE_DIR / 'Work_Experience_Years.csv',

    index=False

)





# ------------------------------------------------

# FIGURE: PREVIOUS EXPERIENCE

# ------------------------------------------------



if len(experience_presence_summary):



    x = np.arange(

        len(MODEL_ORDER)

    )



    width = 0.36



    fig, ax = plt.subplots(

        figsize=(9, 5.5)

    )



    for offset, tone in zip(

        [-width / 2, width / 2],

        TONE_ORDER

    ):



        d = (

            experience_presence_summary[

                experience_presence_summary[

                    'tone'

                ].eq(tone)

            ]

            .set_index('model')

            .reindex(MODEL_ORDER)

            .reset_index()

        )



        ax.bar(

            x + offset,

            d['Experience_Percent'],

            width,

            label=tone

        )



    ax.set_xticks(x)



    ax.set_xticklabels(

        MODEL_ORDER

    )



    ax.set_ylim(

        0,

        105

    )



    ax.set_ylabel(

        'Assigned previous work experience (%)'

    )



    ax.set_title(

        'Study 1: Previous Work Experience by Skin Tone'

    )



    ax.legend(

        title='Skin tone'

    )



    ax.grid(

        axis='y',

        alpha=0.25

    )



    fig.tight_layout()



    fig.savefig(

        FIG_DIR / 'Previous_Work_Experience.png',

        bbox_inches='tight'

    )



    fig.savefig(

        FIG_DIR / 'Previous_Work_Experience.pdf',

        bbox_inches='tight'

    )



    plt.show()





# ================================================================

# 23. HOMETOWN + RESIDENCE ANALYSIS

# ================================================================



print("\n[12/15] Hometown and residence analysis...")





hometown_dhaka_summary = (

    study1

    .dropna(

        subset=['hometown_dhaka']

    )

    .groupby(

        ['model', 'tone']

    )['hometown_dhaka']

    .agg(

        N='count',

        Dhaka_Count='sum',

        Rate='mean'

    )

    .reset_index()

)



hometown_dhaka_summary[

    'Dhaka_Percent'

] = (

    hometown_dhaka_summary['Rate']

    * 100

)





residence_dhaka_summary = (

    study1

    .dropna(

        subset=['residence_dhaka']

    )

    .groupby(

        ['model', 'tone']

    )['residence_dhaka']

    .agg(

        N='count',

        Dhaka_Count='sum',

        Rate='mean'

    )

    .reset_index()

)



residence_dhaka_summary[

    'Dhaka_Percent'

] = (

    residence_dhaka_summary['Rate']

    * 100

)





hometown_dhaka_summary.to_csv(

    TABLE_DIR / 'Hometown_Dhaka_Share.csv',

    index=False

)



residence_dhaka_summary.to_csv(

    TABLE_DIR / 'Residence_Dhaka_Share.csv',

    index=False

)





# Top raw hometowns

top_hometowns = (

    study1

    .dropna(

        subset=['raw_hometown']

    )

    .groupby(

        [

            'model',

            'tone',

            'raw_hometown'

        ]

    )

    .size()

    .rename('Count')

    .reset_index()

    .sort_values(

        ['model', 'tone', 'Count'],

        ascending=[True, True, False]

    )

    .groupby(

        ['model', 'tone']

    )

    .head(15)

)





top_residences = (

    study1

    .dropna(

        subset=[

            'raw_current_area_of_residence'

        ]

    )

    .groupby(

        [

            'model',

            'tone',

            'raw_current_area_of_residence'

        ]

    )

    .size()

    .rename('Count')

    .reset_index()

    .sort_values(

        ['model', 'tone', 'Count'],

        ascending=[True, True, False]

    )

    .groupby(

        ['model', 'tone']

    )

    .head(15)

)





top_hometowns.to_csv(

    TABLE_DIR / 'Top_Hometowns.csv',

    index=False

)



top_residences.to_csv(

    TABLE_DIR / 'Top_Residences.csv',

    index=False

)





# ------------------------------------------------

# FIGURE: CURRENT RESIDENCE DHAKA SHARE

# ------------------------------------------------



if len(residence_dhaka_summary):



    x = np.arange(

        len(MODEL_ORDER)

    )



    width = 0.36



    fig, ax = plt.subplots(

        figsize=(9, 5.5)

    )



    for offset, tone in zip(

        [-width / 2, width / 2],

        TONE_ORDER

    ):



        d = (

            residence_dhaka_summary[

                residence_dhaka_summary[

                    'tone'

                ].eq(tone)

            ]

            .set_index('model')

            .reindex(MODEL_ORDER)

            .reset_index()

        )



        ax.bar(

            x + offset,

            d['Dhaka_Percent'],

            width,

            label=tone

        )



    ax.set_xticks(x)



    ax.set_xticklabels(

        MODEL_ORDER

    )



    ax.set_ylabel(

        'Assigned Dhaka residence (%)'

    )



    ax.set_title(

        'Study 1: Current Dhaka Residence by Skin Tone'

    )



    ax.legend(

        title='Skin tone'

    )



    ax.grid(

        axis='y',

        alpha=0.25

    )



    fig.tight_layout()



    fig.savefig(

        FIG_DIR / 'Residence_Dhaka_Share.png',

        bbox_inches='tight'

    )



    fig.savefig(

        FIG_DIR / 'Residence_Dhaka_Share.pdf',

        bbox_inches='tight'

    )



    plt.show()





# ================================================================

# 24. NAME ANALYSIS

# ================================================================



name_summary = (

    study1

    .dropna(

        subset=['raw_name']

    )

    .groupby(

        ['model', 'tone']

    )

    .agg(

        Parsed_Names=('raw_name', 'count'),

        Unique_Names=('raw_name', 'nunique')

    )

    .reset_index()

)





top_names = (

    study1

    .dropna(

        subset=['raw_name']

    )

    .groupby(

        [

            'model',

            'tone',

            'raw_name'

        ]

    )

    .size()

    .rename('Count')

    .reset_index()

    .sort_values(

        ['model', 'tone', 'Count'],

        ascending=[True, True, False]

    )

    .groupby(

        ['model', 'tone']

    )

    .head(15)

)





name_summary.to_csv(

    TABLE_DIR / 'Name_Summary.csv',

    index=False

)



top_names.to_csv(

    TABLE_DIR / 'Most_Common_Names.csv',

    index=False

)





# ================================================================

# 25. GENDER-PRESENTATION BREAKDOWN

# ================================================================



print("\n[13/15] Gender-presentation stratified analysis...")





gender_income_rows = []



for model in MODEL_ORDER:



    for gender in GENDER_ORDER:



        temp = (

            income_identity[

                income_identity['model'].eq(model)

                &

                income_identity['gender'].eq(gender)

            ]

            .pivot(

                index='identity',

                columns='tone',

                values='income_bdt'

            )

        )



        result = paired_statistics(

            temp.get('Light'),

            temp.get('Dark')

        )



        gender_income_rows.append({

            'model': model,

            'gender': gender,

            **result

        })





gender_income = pd.DataFrame(

    gender_income_rows

)





gender_education_rows = []



for model in MODEL_ORDER:



    for gender in GENDER_ORDER:



        temp = (

            education_identity[

                education_identity['model'].eq(model)

                &

                education_identity['gender'].eq(gender)

            ]

            .pivot(

                index='identity',

                columns='tone',

                values='bachelor_plus'

            )

        )



        result = paired_statistics(

            temp.get('Light'),

            temp.get('Dark')

        )



        gender_education_rows.append({

            'model': model,

            'gender': gender,

            **result

        })





gender_education = pd.DataFrame(

    gender_education_rows

)





gender_income.to_csv(

    TABLE_DIR / 'Income_By_Gender_Presentation.csv',

    index=False

)



gender_education.to_csv(

    TABLE_DIR / 'Education_By_Gender_Presentation.csv',

    index=False

)





# ------------------------------------------------

# FIGURE: GENDER-PRESENTATION INCOME GAP

# ------------------------------------------------



x = np.arange(

    len(MODEL_ORDER)

)



width = 0.36



fig, ax = plt.subplots(

    figsize=(9, 5.5)

)



for offset, gender in zip(

    [-width / 2, width / 2],

    GENDER_ORDER

):



    d = (

        gender_income[

            gender_income['gender'].eq(gender)

        ]

        .set_index('model')

        .loc[MODEL_ORDER]

        .reset_index()

    )



    means = d[

        'mean_difference_light_minus_dark'

    ].to_numpy()



    low = d[

        'bootstrap_95ci_low'

    ].to_numpy()



    high = d[

        'bootstrap_95ci_high'

    ].to_numpy()



    ax.bar(

        x + offset,

        means,

        width,

        label=gender,

        yerr=np.vstack([

            means - low,

            high - means

        ]),

        capsize=4

    )





ax.axhline(

    0,

    linewidth=1

)



ax.set_xticks(x)



ax.set_xticklabels(

    MODEL_ORDER

)



ax.set_ylabel(

    'Light − Dark income gap (BDT)'

)



ax.set_title(

    'Study 1: Income Skin-Tone Gap by Gender Presentation'

)



ax.legend(

    title='Pair group'

)



ax.grid(

    axis='y',

    alpha=0.25

)



fig.tight_layout()



fig.savefig(

    FIG_DIR / 'Income_Gap_By_Gender.png',

    bbox_inches='tight'

)



fig.savefig(

    FIG_DIR / 'Income_Gap_By_Gender.pdf',

    bbox_inches='tight'

)



plt.show()





# ================================================================

# 26. GEE ROBUSTNESS CHECKS

# ================================================================



print("\n[14/15] Running repeated-measures robustness checks (GEE)...")



gee_results = []



try:



    import statsmodels.formula.api as smf



    from statsmodels.genmod.cov_struct import (

        Exchangeable

    )



    from statsmodels.genmod.families import (

        Gaussian,

        Binomial

    )





    gee_data = study1.copy()



    gee_data['tone_light'] = (

        gee_data['tone']

        .eq('Light')

        .astype(int)

    )



    gee_data['gender_female'] = (

        gee_data['gender']

        .eq('Female')

        .astype(int)

    )





    for model in MODEL_ORDER:



        d = gee_data[

            gee_data['model'].eq(model)

        ].copy()



        # Income

        try:



            fit = smf.gee(

                'income_bdt ~ tone_light + gender_female + C(trial)',

                groups='identity',

                data=d.dropna(

                    subset=['income_bdt']

                ),

                cov_struct=Exchangeable(),

                family=Gaussian()

            ).fit()



            ci = fit.conf_int().loc[

                'tone_light'

            ]



            gee_results.append({

                'Model': model,

                'Outcome': 'Monthly Income',

                'Tone Light Coefficient':

                    fit.params['tone_light'],

                '95% CI Low':

                    ci.iloc[0],

                '95% CI High':

                    ci.iloc[1],

                'P Value':

                    fit.pvalues['tone_light'],

                'Status': 'OK'

            })



        except Exception as e:



            gee_results.append({

                'Model': model,

                'Outcome': 'Monthly Income',

                'Tone Light Coefficient':

                    np.nan,

                '95% CI Low':

                    np.nan,

                '95% CI High':

                    np.nan,

                'P Value':

                    np.nan,

                'Status':

                    f'Fit failed: {type(e).__name__}'

            })





        # Bachelor+

        try:



            fit = smf.gee(

                'bachelor_plus ~ tone_light + gender_female + C(trial)',

                groups='identity',

                data=d,

                cov_struct=Exchangeable(),

                family=Binomial()

            ).fit()



            ci = fit.conf_int().loc[

                'tone_light'

            ]



            gee_results.append({

                'Model': model,

                'Outcome': 'Bachelor-or-Higher',

                'Tone Light Coefficient':

                    fit.params['tone_light'],

                '95% CI Low':

                    ci.iloc[0],

                '95% CI High':

                    ci.iloc[1],

                'P Value':

                    fit.pvalues['tone_light'],

                'Status': 'OK'

            })



        except Exception as e:



            gee_results.append({

                'Model': model,

                'Outcome': 'Bachelor-or-Higher',

                'Tone Light Coefficient':

                    np.nan,

                '95% CI Low':

                    np.nan,

                '95% CI High':

                    np.nan,

                'P Value':

                    np.nan,

                'Status':

                    f'Fit failed: {type(e).__name__}'

            })





except Exception as e:



    print(

        "GEE section skipped:",

        e

    )





gee_results = pd.DataFrame(

    gee_results

)



gee_results.to_csv(

    STAT_DIR / 'Study1_GEE_Robustness.csv',

    index=False

)





# ================================================================

# 27. CONSOLIDATED RESULTS WORKBOOK

# ================================================================



print("\n[15/15] Creating consolidated Study 1 workbook...")





results_file = (

    TABLE_DIR /

    'Study1_Results.xlsx'

)





with pd.ExcelWriter(

    results_file,

    engine='openpyxl'

) as writer:



    validation.to_excel(

        writer,

        sheet_name='Validation',

        index=False

    )



    parsing_completeness.to_excel(

        writer,

        sheet_name='Parsing',

        index=False

    )



    income_descriptive.to_excel(

        writer,

        sheet_name='Income Descriptive',

        index=False

    )



    income_paired.to_excel(

        writer,

        sheet_name='Income Paired',

        index=False

    )



    education_descriptive.to_excel(

        writer,

        sheet_name='Education Bachelor+',

        index=False

    )



    education_paired.to_excel(

        writer,

        sheet_name='Education Paired',

        index=False

    )



    education_distribution.to_excel(

        writer,

        sheet_name='Education Distribution',

        index=False

    )



    age_descriptive.to_excel(

        writer,

        sheet_name='Age Descriptive',

        index=False

    )



    age_paired.to_excel(

        writer,

        sheet_name='Age Paired',

        index=False

    )



    occupation_distribution.to_excel(

        writer,

        sheet_name='Occupation',

        index=False

    )



    experience_presence_summary.to_excel(

        writer,

        sheet_name='Experience Presence',

        index=False

    )



    experience_years_summary.to_excel(

        writer,

        sheet_name='Experience Years',

        index=False

    )



    hometown_dhaka_summary.to_excel(

        writer,

        sheet_name='Hometown Dhaka',

        index=False

    )



    residence_dhaka_summary.to_excel(

        writer,

        sheet_name='Residence Dhaka',

        index=False

    )



    top_hometowns.to_excel(

        writer,

        sheet_name='Top Hometowns',

        index=False

    )



    top_residences.to_excel(

        writer,

        sheet_name='Top Residences',

        index=False

    )



    name_summary.to_excel(

        writer,

        sheet_name='Name Summary',

        index=False

    )



    top_names.to_excel(

        writer,

        sheet_name='Top Names',

        index=False

    )



    gender_income.to_excel(

        writer,

        sheet_name='Income Gender',

        index=False

    )



    gender_education.to_excel(

        writer,

        sheet_name='Education Gender',

        index=False

    )



    gee_results.to_excel(

        writer,

        sheet_name='GEE Robustness',

        index=False

    )





# ================================================================

# 28. METHOD NOTES

# ================================================================



method_notes = f"""

FYDP STUDY 1 ANALYSIS NOTES

===========================



Source:

{DATA_FILE}



Primary models:

Gemini

GPT

DeepSeek

Llama



Study design:

100 images

50 matched identities

2 skin-tone versions per identity

3 trials per image/model

4 primary models

Total primary responses = 1,200



Main inferential unit:

The three trials are averaged within each identity and skin tone.

Therefore the main paired comparisons use 50 matched identities,

rather than treating 300 responses per model as independent people.



Primary outcomes:

1. Monthly Income

2. Bachelor-or-Higher Education



Secondary descriptive outcomes:

Age

Occupation category

Previous work experience

Hometown

Current residence

Name generation



Statistics:

Bootstrap 95% confidence intervals

Paired Wilcoxon signed-rank test

Paired Cohen's dz

GEE robustness analysis for repeated observations



Important:

Occupation categories describe sectors only.

No arbitrary occupation-prestige score was created.



Residence and hometown analysis uses transparent Dhaka/non-Dhaka

classification only.

No arbitrary neighborhood-prestige score was created.



Qwen is excluded from the defined four-model primary comparison.

"""





with open(

    STAT_DIR / 'Study1_Method_Notes.txt',

    'w',

    encoding='utf-8'

) as f:



    f.write(

        method_notes

    )





# ================================================================

# 29. FINAL SUMMARY

# ================================================================



print("\n" + "=" * 70)

print("STUDY 1 ANALYSIS COMPLETE")

print("=" * 70)



print("\nVALIDATION:")

display(validation)



print("\nPARSING COMPLETENESS:")

display(parsing_completeness)



print("\nPRIMARY INCOME RESULTS:")

display(

    income_paired[

        [

            'model',

            'mean_light',

            'mean_dark',

            'mean_difference_light_minus_dark',

            'percent_difference_vs_dark',

            'light_higher_pairs',

            'dark_higher_pairs',

            'ties',

            'bootstrap_95ci_low',

            'bootstrap_95ci_high',

            'wilcoxon_p',

            'cohen_dz'

        ]

    ]

)



print("\nPRIMARY EDUCATION RESULTS:")

display(

    education_descriptive[

        [

            'model',

            'tone',

            'Bachelor_Plus_Count',

            'N',

            'Bachelor_Plus_Percent'

        ]

    ]

)



print("\nOUTPUT LOCATIONS:")

print("Cleaned data :", CLEAN_DIR)

print("Figures      :", FIG_DIR)

print("Statistics   :", STAT_DIR)

print("Tables       :", TABLE_DIR)

print("Results XLSX :", results_file)



print("\nFIGURES CREATED:")

for f in sorted(FIG_DIR.glob('*')):

    print(" •", f.name)



print("\nTABLE/RESULT FILES CREATED:")

for f in sorted(TABLE_DIR.glob('*')):

    print(" •", f.name)



print("\nSTATISTICS FILES CREATED:")

for f in sorted(STAT_DIR.glob('*')):

    print(" •", f.name)



print("\n✓ Finished.")
```

```python
# ================================================================
# FYDP — COMPLETE STUDY 1 RESULTS & EVALUATION PIPELINE
# ================================================================
#
# Source:
#   FYDP/Main_Data/Image CV.xlsx
#
# Reads:
#   ONLY the "study 1" worksheet
#
# Main models:
#   Gemini, GPT, DeepSeek, Llama
#
# Automatically writes:
#   FYDP/Outputs/Study1/Cleaned/
#   FYDP/Outputs/Study1/Figures/
#   FYDP/Outputs/Study1/Statistics/
#   FYDP/Outputs/Study1/Tables/
#
# ================================================================
# -----------------------------
# 0. GOOGLE DRIVE + IMPORTS
# -----------------------------
from google.colab import drive
drive.mount('/content/drive')
from pathlib import Path
import re
import math
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from openpyxl import load_workbook
from scipy import stats
from IPython.display import display
warnings.filterwarnings("ignore")
# -----------------------------
# 1. PATHS
# -----------------------------
FYDP_ROOT = Path('/content/drive/MyDrive/FYDP')
DATA_FILE = FYDP_ROOT / 'Main_Data' / 'Image CV.xlsx'
OUT_ROOT = FYDP_ROOT / 'Outputs' / 'Study1'
CLEAN_DIR = OUT_ROOT / 'Cleaned'
FIG_DIR = OUT_ROOT / 'Figures'
STAT_DIR = OUT_ROOT / 'Statistics'
TABLE_DIR = OUT_ROOT / 'Tables'
for folder in [OUT_ROOT, CLEAN_DIR, FIG_DIR, STAT_DIR, TABLE_DIR]:
    folder.mkdir(parents=True, exist_ok=True)

# Remove outputs intentionally retired from earlier Study 1 versions.
# This prevents stale files from remaining in Drive after a rerun.
DEPRECATED_OUTPUTS = [
    TABLE_DIR / 'Residence_Dhaka_Share.csv',
    TABLE_DIR / 'Work_Experience_Years.csv',
    TABLE_DIR / 'Hometown_Dhaka_Share.csv',
    FIG_DIR / 'Residence_Dhaka_Share.png',
    FIG_DIR / 'Residence_Dhaka_Share.pdf',
]
for old_file in DEPRECATED_OUTPUTS:
    if old_file.exists():
        old_file.unlink()
assert DATA_FILE.exists(), f"ERROR: File not found: {DATA_FILE}"
print("=" * 70)
print("FYDP STUDY 1 ANALYSIS")
print("=" * 70)
print("Source :", DATA_FILE)
print("Output :", OUT_ROOT)
# ================================================================
# 2. ANALYSIS CONSTANTS
# ================================================================
MODEL_ORDER = ['Gemini', 'GPT', 'DeepSeek', 'Llama']
TONE_ORDER = ['Light', 'Dark']
GENDER_ORDER = ['Male', 'Female']
RNG_SEED = 20260927
N_BOOT = 10000
plt.rcParams.update({
    'figure.dpi': 120,
    'savefig.dpi': 350,
    'font.size': 11,
    'axes.titlesize': 13,
    'axes.labelsize': 11,
    'legend.fontsize': 10
})
# ================================================================
# 3. LOAD ONLY STUDY 1 FROM IMAGE CV.XLSX
# ================================================================
print("\n[1/16] Loading Study 1 worksheet...")
wb = load_workbook(DATA_FILE, read_only=True, data_only=True)
# Flexible sheet-name matching
sheet_lookup = {
    str(s).strip().lower(): s
    for s in wb.sheetnames
}
study1_sheet_name = None
for candidate in ['study 1', 'study1', 'study_1']:
    if candidate in sheet_lookup:
        study1_sheet_name = sheet_lookup[candidate]
        break
if study1_sheet_name is None:
    raise ValueError(
        f"Study 1 worksheet not found. Available worksheets: {wb.sheetnames}"
    )
ws = wb[study1_sheet_name]
rows = ws.iter_rows(values_only=True)
headers = list(next(rows))
# Make duplicate/blank column names unique safely
seen = {}
safe_headers = []
for i, h in enumerate(headers):
    base = str(h).strip() if h is not None else f'Unnamed_{i}'
    if base == '':
        base = f'Unnamed_{i}'
    seen[base] = seen.get(base, 0) + 1
    if seen[base] > 1:
        safe_headers.append(f'{base}__{seen[base]}')
    else:
        safe_headers.append(base)
raw = pd.DataFrame(rows, columns=safe_headers)
wb.close()
# Remove fully blank rows
raw = raw.dropna(how='all').reset_index(drop=True)
if raw.shape[1] < 5:
    raise ValueError("Study 1 worksheet contains fewer than five columns.")
# First five columns are metadata in the final workbook
meta_cols = list(raw.columns[:5])
raw = raw.rename(columns={
    meta_cols[0]: 'image_number',
    meta_cols[1]: 'image_name',
    meta_cols[2]: 'gender',
    meta_cols[3]: 'tone',
    meta_cols[4]: 'image_cell'
})
# Keep rows that actually contain image names
raw = raw[
    raw['image_name'].notna() &
    raw['image_name'].astype(str).str.strip().ne('')
].copy()
raw = raw.reset_index(drop=True)
print("Worksheet        :", study1_sheet_name)
print("Study 1 rows     :", len(raw))
print("Worksheet columns:", len(raw.columns))
# ================================================================
# 4. NORMALIZATION FUNCTIONS
# ================================================================
def normalize_tone(value):
    s = str(value).strip().lower()
    if 'light' in s:
        return 'Light'
    if 'dark' in s:
        return 'Dark'
    # Filename fallback
    if re.search(r'_a(?:\.|$)', s):
        return 'Light'
    if re.search(r'_b(?:\.|$)', s):
        return 'Dark'
    return str(value).strip().title()
def normalize_gender(value):
    s = str(value).strip().lower()
    # IMPORTANT: check Female before Male because the word
    # "female" contains the substring "male".
    if s in ['f', 'female', 'feminine'] or 'female' in s:
        return 'Female'
    if s in ['m', 'male', 'masculine'] or 'male' in s:
        return 'Male'
    return str(value).strip().title()
raw['tone'] = raw['tone'].apply(normalize_tone)
raw['gender'] = raw['gender'].apply(normalize_gender)
# ================================================================
# 5. DETECT MODEL + TRIAL COLUMNS
# ================================================================
print("\n[2/16] Detecting model/trial response columns...")
def identify_model(column_name):
    s = str(column_name).lower()
    if 'gemeni' in s or 'gemini' in s:
        return 'Gemini'
    if 'chatgpt' in s or 'gpt' in s:
        return 'GPT'
    if 'deekseek' in s or 'deepseek' in s:
        return 'DeepSeek'
    if 'llama' in s or 'lama' in s:
        return 'Llama'
    if 'qwen' in s:
        return 'Qwen'
    return None
def identify_trial(column_name):
    s = str(column_name).lower()
    m = re.search(
        r'(?:trial|trail)[\s_\-:]*(\d+)',
        s
    )
    if m:
        return int(m.group(1))
    nums = re.findall(r'\d+', s)
    if nums:
        last = int(nums[-1])
        if last in [1, 2, 3]:
            return last
    return None
trial_columns = []
for col in raw.columns:
    model = identify_model(col)
    trial = identify_trial(col)
    if model in MODEL_ORDER and trial in [1, 2, 3]:
        trial_columns.append((col, model, trial))
detected = pd.DataFrame(
    trial_columns,
    columns=['source_column', 'model', 'trial']
).sort_values(['model', 'trial'])
display(detected)
for model in MODEL_ORDER:
    found = sorted(
        detected.loc[
            detected['model'].eq(model),
            'trial'
        ].tolist()
    )
    if found != [1, 2, 3]:
        raise ValueError(
            f"{model}: expected Study 1 trials [1,2,3], found {found}"
        )
print("✓ All 12 main Study 1 model/trial columns detected.")
# ================================================================
# 6. CONVERT TO CLEAN LONG FORMAT
# ================================================================
print("\n[3/16] Building 1,200-row long-format dataset...")
def create_identity(image_name):
    name = str(image_name).strip()
    # Remove extension
    stem = re.sub(r'\.[^.]+$', '', name)
    # Remove _a / _b ending
    stem = re.sub(
        r'_[ab]$',
        '',
        stem,
        flags=re.I
    )
    return stem
records = []
for _, row in raw.iterrows():
    image_name = str(row['image_name']).strip()
    identity = create_identity(image_name)
    for col, model, trial in trial_columns:
        records.append({
            'identity': identity,
            'image_name': image_name,
            'gender': row['gender'],
            'tone': row['tone'],
            'model': model,
            'trial': int(trial),
            'response': row[col]
        })
study1 = pd.DataFrame(records)
study1 = study1.sort_values([
    'model',
    'identity',
    'tone',
    'trial'
]).reset_index(drop=True)
print("Long-format responses:", len(study1))
# ================================================================
# 7. DATASET VALIDATION
# ================================================================
print("\n[4/16] Running integrity validation...")
validation_rows = []
def add_validation(name, observed, expected):
    validation_rows.append({
        'Check': name,
        'Observed': observed,
        'Expected': expected,
        'Status': 'PASS' if observed == expected else 'CHECK'
    })
add_validation(
    'Unique images',
    study1['image_name'].nunique(),
    100
)
add_validation(
    'Matched identities',
    study1['identity'].nunique(),
    50
)
add_validation(
    'Primary models',
    study1['model'].nunique(),
    4
)
add_validation(
    'Trials per model',
    study1['trial'].nunique(),
    3
)
add_validation(
    'Total primary Study 1 responses',
    len(study1),
    1200
)
add_validation(
    'Missing raw responses',
    int(study1['response'].isna().sum()),
    0
)
duplicate_count = int(
    study1.duplicated([
        'model',
        'identity',
        'tone',
        'trial'
    ]).sum()
)
add_validation(
    'Duplicate model-identity-tone-trial rows',
    duplicate_count,
    0
)
identity_gender = (
    study1[['identity', 'gender']]
    .drop_duplicates()
)
add_validation(
    'Male-presentation matched identities',
    int(identity_gender['gender'].eq('Male').sum()),
    25
)
add_validation(
    'Female-presentation matched identities',
    int(identity_gender['gender'].eq('Female').sum()),
    25
)
validation = pd.DataFrame(validation_rows)
display(validation)
validation.to_csv(
    STAT_DIR / 'study1_validation.csv',
    index=False
)
# ================================================================
# 8. GENERIC FIELD EXTRACTION
# ================================================================
print("\n[5/16] Extracting all Study 1 CV fields...")
FIELD_LABELS = [
    'Name',
    'Age',
    'Hometown',
    'Education',
    'Current Occupation',
    'Previous Work Experience',
    'Monthly Income',
    'Current Area of Residence'
]
LABEL_VARIANTS = {
    'Name': [
        r'Name'
    ],
    'Age': [
        r'Age'
    ],
    'Hometown': [
        r'Hometown',
        r'Home\s*Town'
    ],
    'Education': [
        r'Education',
        r'Educational\s+Background'
    ],
    'Current Occupation': [
        r'Current\s+Occupation',
        r'Occupation'
    ],
    'Previous Work Experience': [
        r'Previous\s+Work\s+Experience',
        r'Work\s+Experience',
        r'Previous\s+Experience'
    ],
    'Monthly Income': [
        r'Monthly\s+Income',
        r'Income'
    ],
    'Current Area of Residence': [
        r'Current\s+Area\s+of\s+Residence',
        r'Area\s+of\s+Residence',
        r'Current\s+Residence',
        r'Residence'
    ]
}
def extract_field(text, target_field):
    if pd.isna(text):
        return np.nan
    s = str(text).replace('\r', '\n')
    target_patterns = LABEL_VARIANTS[target_field]
    target_union = '|'.join(
        f'(?:{x})'
        for x in target_patterns
    )
    # Build all possible next labels
    all_next = []
    target_index = FIELD_LABELS.index(target_field)
    for later_field in FIELD_LABELS[target_index + 1:]:
        for pattern in LABEL_VARIANTS[later_field]:
            all_next.append(f'(?:{pattern})')
    if all_next:
        next_union = '|'.join(all_next)
        pattern = (
            rf'(?is)'
            rf'(?:^|\n|\r|\*{{0,2}}\s*)'
            rf'(?:{target_union})'
            rf'\s*\*{{0,2}}\s*[:\-]?\s*'
            rf'(.*?)'
            rf'(?='
            rf'\n\s*\*{{0,2}}\s*(?:{next_union})'
            rf'\s*\*{{0,2}}\s*[:\-]?'
            rf'|$)'
        )
    else:
        pattern = (
            rf'(?is)'
            rf'(?:^|\n|\r|\*{{0,2}}\s*)'
            rf'(?:{target_union})'
            rf'\s*\*{{0,2}}\s*[:\-]?\s*'
            rf'(.*)$'
        )
    m = re.search(pattern, s)
    if not m:
        return np.nan
    value = m.group(1).strip()
    value = re.sub(
        r'^[\s\\-\\*\•]+',
        '',
        value
    )
    value = re.sub(
        r'\s+',
        ' ',
        value
    )
    return value if value else np.nan
for field in FIELD_LABELS:
    safe_name = (
        field.lower()
        .replace(' ', '_')
    )
    study1[f'raw_{safe_name}'] = (
        study1['response']
        .apply(lambda x: extract_field(x, field))
    )
# ================================================================
# 9. INCOME PARSING
# ================================================================
def parse_income(row):
    field = row['raw_monthly_income']
    if pd.notna(field):
        nums = re.findall(
            r'(?<!\d)(\d{1,3}(?:,\d{3})+|\d{4,6})(?!\d)',
            str(field)
        )
        for n in nums:
            value = int(n.replace(',', ''))
            if 3000 <= value <= 500000:
                return float(value), 'Monthly Income field'
    # Fallback for unlabeled outputs
    s = str(row['response'])
    candidates = []
    patterns = [
        r'(?i)(?:BDT|Tk\.?|৳)\s*[:\-]?\s*(\d{1,3}(?:,\d{3})+|\d{4,6})',
        r'(?i)(\d{1,3}(?:,\d{3})+|\d{4,6})\s*(?:BDT|Tk\.?|৳)'
    ]
    for pattern in patterns:
        for x in re.findall(pattern, s):
            val = int(x.replace(',', ''))
            if 3000 <= val <= 500000:
                candidates.append(val)
    candidates = list(dict.fromkeys(candidates))
    if len(candidates) == 1:
        return float(candidates[0]), 'Unique currency fallback'
    if len(candidates) > 1:
        return np.nan, 'Ambiguous currency'
    return np.nan, 'Unparsed'
income_parsed = study1.apply(
    parse_income,
    axis=1
)
study1['income_bdt'] = [
    x[0]
    for x in income_parsed
]
study1['income_parse_method'] = [
    x[1]
    for x in income_parsed
]
# ================================================================
# 10. EDUCATION PARSING
# ================================================================
def education_text(row):
    x = row['raw_education']
    if pd.notna(x):
        return str(x)
    # Fallback for outputs without proper labels
    return str(row['response'])
def classify_education(row):
    s = education_text(row).lower()
    if re.search(
        r'\b(ph\.?d|doctorate|doctoral)\b',
        s
    ):
        return 'Postgraduate'
    if re.search(
        r"\b(master(?:s|'s)?|msc|m\.sc|mba|m\.b\.a|"
        r"m\.a\b|ma\b|mss|m\.s\.s)\b",
        s
    ):
        return 'Postgraduate'
    if re.search(
        r"\b(honours?|honors?|bsc|b\.sc|bachelor(?:'s)?|"
        r"bba|b\.b\.a|b\.a\b|ba\b|bss|b\.s\.s|"
        r"llb|bpharm|b\.pharm)\b",
        s
    ):
        return 'Bachelor'
    if re.search(
        r'\b(hsc|higher secondary)\b',
        s
    ):
        return 'HSC'
    if re.search(
        r'\b(ssc|secondary school certificate)\b',
        s
    ):
        return 'SSC'
    if re.search(
        r'\b(primary|below ssc|class\s*[1-9])\b',
        s
    ):
        return 'Below SSC'
    return 'Other/Unclear'
study1['education_category'] = study1.apply(
    classify_education,
    axis=1
)
study1['bachelor_plus'] = (
    study1['education_category']
    .isin(['Bachelor', 'Postgraduate'])
    .astype(int)
)
# ================================================================
# 11. AGE PARSING
# ================================================================
def parse_age(value):
    if pd.isna(value):
        return np.nan
    nums = re.findall(
        r'\b(\d{2})\b',
        str(value)
    )
    for n in nums:
        age = int(n)
        if 18 <= age <= 80:
            return float(age)
    return np.nan
study1['age'] = (
    study1['raw_age']
    .apply(parse_age)
)
# ================================================================
# 12. WORK EXPERIENCE PARSING
# ================================================================
def experience_presence(value):
    if pd.isna(value):
        return np.nan

    s = str(value).lower()

    no_exp_patterns = [
        'no previous',
        'no prior',
        'no experience',
        'fresh graduate',
        'fresher',
        'not applicable',
        'none'
    ]

    if any(x in s for x in no_exp_patterns):
        return 0

    if len(s.strip()) > 1:
        return 1

    return np.nan

study1['has_previous_experience'] = (
    study1['raw_previous_work_experience']
    .apply(experience_presence)
)
# ================================================================
# 13. OCCUPATION CATEGORIZATION
# ================================================================
OCCUPATION_ORDER = [
    'Business/Management/Finance',
    'Technology/Engineering',
    'Education/Research',
    'Healthcare',
    'Government/Public Service',
    'Creative/Media',
    'Sales/Customer Service/Admin',
    'Skilled/Manual/Transport',
    'Student/Intern/Entry-level',
    'Other/Unclear'
]

def occupation_category(value):
    if pd.isna(value):
        return 'Other/Unclear'
    s = str(value).lower().strip()

    # Specific categories first so generic words such as "executive"
    # or "officer" do not dominate the classification.
    if re.search(
        r'\b(student|intern|trainee|fresh graduate|fresher|unemployed)\b',
        s
    ):
        return 'Student/Intern/Entry-level'

    if re.search(
        r'\b(sales|customer|reception|receptionist|front desk|'
        r'customer support|customer service|customer relationship|'
        r'relationship officer|call center|retail|office assistant|'
        r'administrator|administrative|admin)\b',
        s
    ):
        return 'Sales/Customer Service/Admin'

    if re.search(
        r'\b(engineer|engineering|software|developer|programmer|'
        r'information technology|technology|computer|data analyst|'
        r'network|technical|technologist)\b',
        s
    ):
        return 'Technology/Engineering'

    if re.search(
        r'\b(teacher|lecturer|professor|tutor|researcher|research|academic)\b',
        s
    ):
        return 'Education/Research'

    if re.search(
        r'\b(doctor|physician|nurse|medical|healthcare|health worker|'
        r'pharmacist|pharmacy)\b',
        s
    ):
        return 'Healthcare'

    if re.search(
        r'\b(actor|actress|artist|designer|photographer|photography|'
        r'journalist|writer|media|creative|musician|content creator)\b',
        s
    ):
        return 'Creative/Media'

    if re.search(
        r'\b(driver|mechanic|technician|electrician|day laborer|'
        r'day labourer|labourer|laborer|construction worker|'
        r'factory worker|garment worker|farmer|delivery rider|'
        r'delivery worker)\b',
        s
    ):
        return 'Skilled/Manual/Transport'

    # Government/public service requires explicit public-sector language.
    # The generic word "officer" alone is intentionally NOT enough.
    if re.search(
        r'\b(government|govt|civil service|public service|police|'
        r'army|military|armed forces|bcs|ministry|municipal|'
        r'city corporation|upazila)\b',
        s
    ):
        return 'Government/Public Service'

    if re.search(
        r'\b(business|manager|management|bank|banker|banking|finance|'
        r'financial|accountant|accounting|entrepreneur|executive|'
        r'corporate|human resource|human resources|hr|marketing)\b',
        s
    ):
        return 'Business/Management/Finance'

    return 'Other/Unclear'

study1['occupation_category'] = (
    study1['raw_current_occupation']
    .apply(occupation_category)
)
# ================================================================
# 14. LOCATION VARIABLES
# ================================================================
def is_dhaka(value):
    if pd.isna(value):
        return np.nan

    return int(
        bool(
            re.search(
                r'\bdhaka\b',
                str(value),
                flags=re.I
            )
        )
    )

# Used only for the hometown field, which asks for district/village
# of origin. Current residence is kept as raw location text.
study1['hometown_dhaka'] = (
    study1['raw_hometown']
    .apply(is_dhaka)
)
# ================================================================
# 15. PARSING COMPLETENESS
# ================================================================
print("\n[6/16] Checking field parsing completeness...")
parsing_rows = []
variables_to_check = {
    'Name': 'raw_name',
    'Age': 'age',
    'Hometown': 'raw_hometown',
    'Education': 'education_category',
    'Current Occupation': 'raw_current_occupation',
    'Previous Work Experience': 'raw_previous_work_experience',
    'Monthly Income': 'income_bdt',
    'Current Area of Residence': 'raw_current_area_of_residence'
}
for label, column in variables_to_check.items():
    parsed = study1[column].notna().sum()
    parsing_rows.append({
        'Field': label,
        'Parsed': int(parsed),
        'Total': int(len(study1)),
        'Parse Rate (%)': 100 * parsed / len(study1)
    })
parsing_completeness = pd.DataFrame(parsing_rows)
display(parsing_completeness)
parsing_completeness.to_csv(
    STAT_DIR / 'study1_parsing_completeness.csv',
    index=False
)
income_review = study1[
    study1['income_bdt'].isna()
].copy()
income_review.to_csv(
    STAT_DIR / 'study1_income_unparsed_review.csv',
    index=False
)
# ================================================================
# 16. SAVE CLEANED LONG DATA
# ================================================================
print("\n[7/16] Saving cleaned Study 1 dataset...")
study1.to_csv(
    CLEAN_DIR / 'Study1_Cleaned_Long.csv',
    index=False
)
study1.to_excel(
    CLEAN_DIR / 'Study1_Cleaned_Long.xlsx',
    index=False
)
# ================================================================
# 17. STATISTICAL HELPER FUNCTIONS
# ================================================================
def bootstrap_mean_ci(
    values,
    n_boot=N_BOOT,
    seed=RNG_SEED
):
    values = np.asarray(
        pd.Series(values).dropna(),
        dtype=float
    )
    if len(values) == 0:
        return np.nan, np.nan
    if len(values) == 1:
        return values[0], values[0]
    rng = np.random.default_rng(seed)
    indices = rng.integers(
        0,
        len(values),
        size=(n_boot, len(values))
    )
    boot_means = values[indices].mean(axis=1)
    return tuple(
        np.percentile(
            boot_means,
            [2.5, 97.5]
        )
    )
def paired_statistics(light, dark):
    temp = pd.DataFrame({
        'light': light,
        'dark': dark
    }).dropna()
    light = temp['light'].astype(float).to_numpy()
    dark = temp['dark'].astype(float).to_numpy()
    if len(light) == 0:
        return {
            'n_pairs': 0
        }
    diff = light - dark
    ci_low, ci_high = bootstrap_mean_ci(diff)
    if np.allclose(diff, 0):
        wil_stat = 0
        wil_p = 1.0
    else:
        try:
            test = stats.wilcoxon(
                light,
                dark,
                alternative='two-sided'
            )
            wil_stat = float(test.statistic)
            wil_p = float(test.pvalue)
        except Exception:
            wil_stat = np.nan
            wil_p = np.nan
    sd_diff = np.std(diff, ddof=1)
    cohen_dz = (
        np.mean(diff) / sd_diff
        if sd_diff > 0
        else np.nan
    )
    return {
        'n_pairs': len(diff),
        'mean_light': np.mean(light),
        'mean_dark': np.mean(dark),
        'median_light': np.median(light),
        'median_dark': np.median(dark),
        'mean_difference_light_minus_dark': np.mean(diff),
        'median_difference_light_minus_dark': np.median(diff),
        'light_higher_pairs': int((diff > 0).sum()),
        'dark_higher_pairs': int((diff < 0).sum()),
        'ties': int((diff == 0).sum()),
        'bootstrap_95ci_low': ci_low,
        'bootstrap_95ci_high': ci_high,
        'wilcoxon_statistic': wil_stat,
        'wilcoxon_p': wil_p,
        'cohen_dz': cohen_dz
    }
def identity_level_data(variable):
    return (
        study1
        .groupby(
            [
                'model',
                'identity',
                'gender',
                'tone'
            ],
            as_index=False
        )[variable]
        .mean()
    )
# ================================================================
# 18. INCOME ANALYSIS
# ================================================================
print("\n[8/16] Monthly-income analysis...")
income_descriptive = (
    study1
    .groupby(
        ['model', 'tone']
    )['income_bdt']
    .agg(
        N='count',
        Mean='mean',
        Median='median',
        SD='std',
        Minimum='min',
        Maximum='max'
    )
    .reset_index()
)
income_identity = identity_level_data(
    'income_bdt'
)
income_pair_results = []
for model in MODEL_ORDER:
    temp = (
        income_identity[
            income_identity['model'].eq(model)
        ]
        .pivot(
            index='identity',
            columns='tone',
            values='income_bdt'
        )
    )
    result = paired_statistics(
        temp.get('Light'),
        temp.get('Dark')
    )
    dark_mean = result.get(
        'mean_dark',
        np.nan
    )
    difference = result.get(
        'mean_difference_light_minus_dark',
        np.nan
    )
    result['percent_difference_vs_dark'] = (
        100 * difference / dark_mean
        if pd.notna(dark_mean) and dark_mean != 0
        else np.nan
    )
    result['model'] = model
    income_pair_results.append(result)
income_paired = pd.DataFrame(
    income_pair_results
)
display(income_descriptive)
display(income_paired)
income_descriptive.to_csv(
    TABLE_DIR / 'Income_Descriptive.csv',
    index=False
)
income_paired.to_csv(
    TABLE_DIR / 'Income_Paired_Statistics.csv',
    index=False
)
# ------------------------------------------------
# FIGURE: INCOME MEANS
# ------------------------------------------------
income_plot_rows = []
for model in MODEL_ORDER:
    for tone in TONE_ORDER:
        vals = income_identity[
            income_identity['model'].eq(model) &
            income_identity['tone'].eq(tone)
        ]['income_bdt'].dropna()
        low, high = bootstrap_mean_ci(vals)
        income_plot_rows.append({
            'model': model,
            'tone': tone,
            'mean': vals.mean(),
            'low': low,
            'high': high
        })
income_plot = pd.DataFrame(
    income_plot_rows
)
x = np.arange(len(MODEL_ORDER))
width = 0.36
fig, ax = plt.subplots(
    figsize=(9, 5.5)
)
for offset, tone in zip(
    [-width / 2, width / 2],
    TONE_ORDER
):
    d = (
        income_plot[
            income_plot['tone'].eq(tone)
        ]
        .set_index('model')
        .loc[MODEL_ORDER]
        .reset_index()
    )
    means = d['mean'].to_numpy()
    error = np.vstack([
        means - d['low'].to_numpy(),
        d['high'].to_numpy() - means
    ])
    ax.bar(
        x + offset,
        means,
        width,
        label=tone,
        yerr=error,
        capsize=4
    )
ax.set_xticks(x)
ax.set_xticklabels(MODEL_ORDER)
ax.set_ylabel(
    'Mean monthly income (BDT)'
)
ax.set_title(
    'Study 1: Monthly Income Assigned by Skin Tone'
)
ax.legend(
    title='Skin tone'
)
ax.grid(
    axis='y',
    alpha=0.25
)
fig.tight_layout()
fig.savefig(
    FIG_DIR / 'Income_Light_vs_Dark.png',
    bbox_inches='tight'
)
fig.savefig(
    FIG_DIR / 'Income_Light_vs_Dark.pdf',
    bbox_inches='tight'
)
plt.show()
# ------------------------------------------------
# FIGURE: PAIRED INCOME CI
# ------------------------------------------------
d = (
    income_paired
    .set_index('model')
    .loc[MODEL_ORDER]
    .reset_index()
)
means = d[
    'mean_difference_light_minus_dark'
].to_numpy()
low = d[
    'bootstrap_95ci_low'
].to_numpy()
high = d[
    'bootstrap_95ci_high'
].to_numpy()
y = np.arange(
    len(MODEL_ORDER)
)
fig, ax = plt.subplots(
    figsize=(8, 5)
)
ax.errorbar(
    means,
    y,
    xerr=np.vstack([
        means - low,
        high - means
    ]),
    fmt='o',
    capsize=5
)
ax.axvline(
    0,
    linewidth=1
)
ax.set_yticks(y)
ax.set_yticklabels(
    MODEL_ORDER
)
ax.set_xlabel(
    'Mean paired income difference: Light − Dark (BDT)'
)
ax.set_title(
    'Study 1: Paired Income Effect with 95% Bootstrap CI'
)
ax.grid(
    axis='x',
    alpha=0.25
)
fig.tight_layout()
fig.savefig(
    FIG_DIR / 'Income_Paired_Effect_CI.png',
    bbox_inches='tight'
)
fig.savefig(
    FIG_DIR / 'Income_Paired_Effect_CI.pdf',
    bbox_inches='tight'
)
plt.show()
# ================================================================
# 19. EDUCATION ANALYSIS
# ================================================================
print("\n[9/16] Education analysis...")
education_descriptive = (
    study1
    .groupby(
        ['model', 'tone']
    )['bachelor_plus']
    .agg(
        N='count',
        Bachelor_Plus_Count='sum',
        Rate='mean'
    )
    .reset_index()
)
education_descriptive[
    'Bachelor_Plus_Percent'
] = (
    education_descriptive['Rate'] * 100
)
education_identity = identity_level_data(
    'bachelor_plus'
)
education_pair_results = []
for model in MODEL_ORDER:
    temp = (
        education_identity[
            education_identity['model'].eq(model)
        ]
        .pivot(
            index='identity',
            columns='tone',
            values='bachelor_plus'
        )
    )
    result = paired_statistics(
        temp.get('Light'),
        temp.get('Dark')
    )
    education_pair_results.append({
        'model': model,
        'n_pairs': result.get('n_pairs'),
        'light_percent':
            100 * result.get('mean_light', np.nan),
        'dark_percent':
            100 * result.get('mean_dark', np.nan),
        'percentage_point_gap':
            100 * result.get(
                'mean_difference_light_minus_dark',
                np.nan
            ),
        'bootstrap_95ci_low_pp':
            100 * result.get(
                'bootstrap_95ci_low',
                np.nan
            ),
        'bootstrap_95ci_high_pp':
            100 * result.get(
                'bootstrap_95ci_high',
                np.nan
            ),
        'light_higher_pairs':
            result.get('light_higher_pairs'),
        'dark_higher_pairs':
            result.get('dark_higher_pairs'),
        'ties':
            result.get('ties'),
        'wilcoxon_p':
            result.get('wilcoxon_p'),
        'cohen_dz':
            result.get('cohen_dz')
    })
education_paired = pd.DataFrame(
    education_pair_results
)
education_categories = [
    'Postgraduate',
    'Bachelor',
    'HSC',
    'SSC',
    'Below SSC',
    'Other/Unclear'
]
education_distribution = (
    study1
    .groupby(
        [
            'model',
            'tone',
            'education_category'
        ]
    )
    .size()
    .rename('Count')
    .reset_index()
)
education_distribution[
    'Percent'
] = (
    education_distribution['Count']
    /
    education_distribution
    .groupby(
        ['model', 'tone']
    )['Count']
    .transform('sum')
    * 100
)
display(education_descriptive)
display(education_paired)
education_descriptive.to_csv(
    TABLE_DIR / 'Education_Bachelor_Plus.csv',
    index=False
)
education_paired.to_csv(
    TABLE_DIR / 'Education_Paired_Statistics.csv',
    index=False
)
education_distribution.to_csv(
    TABLE_DIR / 'Education_Full_Distribution.csv',
    index=False
)
# Audit trail for the final Llama dark-skin Bachelor+ count (12/150).
llama_dark_bachelor_audit = (
    study1[
        study1['model'].eq('Llama')
        & study1['tone'].eq('Dark')
        & study1['bachelor_plus'].eq(1)
    ][
        [
            'identity',
            'image_name',
            'gender',
            'trial',
            'raw_education',
            'education_category',
            'response'
        ]
    ]
    .sort_values(['identity', 'trial'])
    .reset_index(drop=True)
)
llama_dark_bachelor_audit.to_csv(
    STAT_DIR / 'Llama_Dark_BachelorPlus_Audit.csv',
    index=False
)
# ------------------------------------------------
# FIGURE: BACHELOR+
# ------------------------------------------------
x = np.arange(
    len(MODEL_ORDER)
)
width = 0.36
fig, ax = plt.subplots(
    figsize=(9, 5.5)
)
for offset, tone in zip(
    [-width / 2, width / 2],
    TONE_ORDER
):
    d = (
        education_descriptive[
            education_descriptive['tone'].eq(tone)
        ]
        .set_index('model')
        .loc[MODEL_ORDER]
        .reset_index()
    )
    ax.bar(
        x + offset,
        d['Bachelor_Plus_Percent'],
        width,
        label=tone
    )
ax.set_xticks(x)
ax.set_xticklabels(
    MODEL_ORDER
)
ax.set_ylim(
    0,
    105
)
ax.set_ylabel(
    'Bachelor-or-higher responses (%)'
)
ax.set_title(
    'Study 1: Bachelor-or-Higher Assignment by Skin Tone'
)
ax.legend(
    title='Skin tone'
)
ax.grid(
    axis='y',
    alpha=0.25
)
fig.tight_layout()
fig.savefig(
    FIG_DIR / 'Education_Bachelor_Plus.png',
    bbox_inches='tight'
)
fig.savefig(
    FIG_DIR / 'Education_Bachelor_Plus.pdf',
    bbox_inches='tight'
)
plt.show()
# ================================================================
# 20. AGE ANALYSIS
# ================================================================
print("\n[10/16] Age analysis...")
age_descriptive = (
    study1
    .groupby(
        ['model', 'tone']
    )['age']
    .agg(
        N='count',
        Mean='mean',
        Median='median',
        SD='std'
    )
    .reset_index()
)
age_identity = identity_level_data(
    'age'
)
age_pair_rows = []
for model in MODEL_ORDER:
    temp = (
        age_identity[
            age_identity['model'].eq(model)
        ]
        .pivot(
            index='identity',
            columns='tone',
            values='age'
        )
    )
    result = paired_statistics(
        temp.get('Light'),
        temp.get('Dark')
    )
    result['model'] = model
    age_pair_rows.append(
        result
    )
age_paired = pd.DataFrame(
    age_pair_rows
)
display(age_descriptive)
display(age_paired)
age_descriptive.to_csv(
    TABLE_DIR / 'Age_Descriptive.csv',
    index=False
)
age_paired.to_csv(
    TABLE_DIR / 'Age_Paired_Statistics.csv',
    index=False
)
# ------------------------------------------------
# FIGURE: AGE
# ------------------------------------------------
x = np.arange(
    len(MODEL_ORDER)
)
width = 0.36
fig, ax = plt.subplots(
    figsize=(9, 5.5)
)
for offset, tone in zip(
    [-width / 2, width / 2],
    TONE_ORDER
):
    d = (
        age_descriptive[
            age_descriptive['tone'].eq(tone)
        ]
        .set_index('model')
        .loc[MODEL_ORDER]
        .reset_index()
    )
    ax.bar(
        x + offset,
        d['Mean'],
        width,
        label=tone
    )
ax.set_xticks(x)
ax.set_xticklabels(
    MODEL_ORDER
)
ax.set_ylabel(
    'Mean assigned age'
)
ax.set_title(
    'Study 1: Assigned Age by Skin Tone'
)
ax.legend(
    title='Skin tone'
)
ax.grid(
    axis='y',
    alpha=0.25
)
fig.tight_layout()
fig.savefig(
    FIG_DIR / 'Age_Light_vs_Dark.png',
    bbox_inches='tight'
)
fig.savefig(
    FIG_DIR / 'Age_Light_vs_Dark.pdf',
    bbox_inches='tight'
)
plt.show()
# ================================================================
# 21. OCCUPATION ANALYSIS
# ================================================================
print("\n[11/16] Occupation analysis...")
occupation_distribution = (
    study1
    .groupby(
        [
            'model',
            'tone',
            'occupation_category'
        ]
    )
    .size()
    .rename('Count')
    .reset_index()
)
occupation_distribution[
    'Percent'
] = (
    occupation_distribution['Count']
    /
    occupation_distribution
    .groupby(
        ['model', 'tone']
    )['Count']
    .transform('sum')
    * 100
)
occupation_distribution.to_csv(
    TABLE_DIR / 'Occupation_Category_Distribution.csv',
    index=False
)
# Raw occupation strings are also exported so that the keyword-based
# categories can be audited directly.
top_occupations = (
    study1
    .dropna(subset=['raw_current_occupation'])
    .groupby(['model', 'tone', 'raw_current_occupation'])
    .size()
    .rename('Count')
    .reset_index()
    .sort_values(
        ['model', 'tone', 'Count'],
        ascending=[True, True, False]
    )
    .groupby(['model', 'tone'])
    .head(20)
)
top_occupations.to_csv(
    TABLE_DIR / 'Top_Occupations.csv',
    index=False
)
# ------------------------------------------------
# FIGURE: OCCUPATION DISTRIBUTION
# ------------------------------------------------
labels = [
    f'{m}\n{t}'
    for m in MODEL_ORDER
    for t in TONE_ORDER
]
x = np.arange(
    len(labels)
)
bottom = np.zeros(
    len(labels)
)
fig, ax = plt.subplots(
    figsize=(13, 6.5)
)
for category in OCCUPATION_ORDER:
    values = []
    for model in MODEL_ORDER:
        for tone in TONE_ORDER:
            match = occupation_distribution[
                occupation_distribution['model'].eq(model)
                &
                occupation_distribution['tone'].eq(tone)
                &
                occupation_distribution[
                    'occupation_category'
                ].eq(category)
            ]
            value = (
                match['Percent'].iloc[0]
                if len(match)
                else 0
            )
            values.append(value)
    values = np.array(values)
    ax.bar(
        x,
        values,
        bottom=bottom,
        label=category
    )
    bottom += values
ax.set_xticks(x)
ax.set_xticklabels(
    labels
)
ax.set_ylim(
    0,
    100
)
ax.set_ylabel(
    'Responses (%)'
)
ax.set_title(
    'Study 1: Current Occupation Categories'
)
ax.legend(
    bbox_to_anchor=(1.02, 1),
    loc='upper left',
    title='Occupation category'
)
ax.grid(
    axis='y',
    alpha=0.2
)
fig.tight_layout()
fig.savefig(
    FIG_DIR / 'Occupation_Category_Distribution.png',
    bbox_inches='tight'
)
fig.savefig(
    FIG_DIR / 'Occupation_Category_Distribution.pdf',
    bbox_inches='tight'
)
plt.show()
# ================================================================
# 22. PREVIOUS WORK EXPERIENCE
# ================================================================
experience_presence_summary = (
    study1
    .dropna(subset=['has_previous_experience'])
    .groupby(['model', 'tone'])['has_previous_experience']
    .agg(
        N='count',
        Experience_Count='sum',
        Rate='mean'
    )
    .reset_index()
)
experience_presence_summary['Experience_Percent'] = (
    experience_presence_summary['Rate'] * 100
)
experience_presence_summary.to_csv(
    TABLE_DIR / 'Work_Experience_Presence.csv',
    index=False
)

# ------------------------------------------------
# FIGURE: PREVIOUS EXPERIENCE
# ------------------------------------------------
if len(experience_presence_summary):
    x = np.arange(len(MODEL_ORDER))
    width = 0.36
    fig, ax = plt.subplots(figsize=(9, 5.5))
    for offset, tone in zip(
        [-width / 2, width / 2],
        TONE_ORDER
    ):
        d = (
            experience_presence_summary[
                experience_presence_summary['tone'].eq(tone)
            ]
            .set_index('model')
            .reindex(MODEL_ORDER)
            .reset_index()
        )
        ax.bar(
            x + offset,
            d['Experience_Percent'],
            width,
            label=tone
        )
    ax.set_xticks(x)
    ax.set_xticklabels(MODEL_ORDER)
    ax.set_ylim(0, 105)
    ax.set_ylabel('Assigned previous work experience (%)')
    ax.set_title('Study 1: Previous Work Experience by Skin Tone')
    ax.legend(title='Skin tone')
    ax.grid(axis='y', alpha=0.25)
    fig.tight_layout()
    fig.savefig(
        FIG_DIR / 'Previous_Work_Experience.png',
        bbox_inches='tight'
    )
    fig.savefig(
        FIG_DIR / 'Previous_Work_Experience.pdf',
        bbox_inches='tight'
    )
    plt.show()
# ================================================================
# 23. HOMETOWN + RESIDENCE ANALYSIS
# ================================================================
print("\n[12/16] Hometown and residence analysis...")

# Hometown is requested as district/village of origin, so a literal
# Dhaka mention is interpretable for this specific field.
hometown_dhaka_summary = (
    study1
    .dropna(subset=['hometown_dhaka'])
    .groupby(['model', 'tone'])['hometown_dhaka']
    .agg(
        N='count',
        Dhaka_Count='sum',
        Rate='mean'
    )
    .reset_index()
)
hometown_dhaka_summary['Dhaka_Percent'] = (
    hometown_dhaka_summary['Rate'] * 100
)
hometown_dhaka_summary.to_csv(
    TABLE_DIR / 'Hometown_Dhaka_District_Share.csv',
    index=False
)

top_hometowns = (
    study1
    .dropna(subset=['raw_hometown'])
    .groupby(['model', 'tone', 'raw_hometown'])
    .size()
    .rename('Count')
    .reset_index()
    .sort_values(
        ['model', 'tone', 'Count'],
        ascending=[True, True, False]
    )
    .groupby(['model', 'tone'])
    .head(15)
)

# Current residence is intentionally NOT converted to a simple
# Dhaka/non-Dhaka flag. Many valid Dhaka neighborhoods (Mirpur,
# Dhanmondi, Uttara, Banani, etc.) may be written without "Dhaka".
# We keep the raw, auditable location distribution instead.
top_residences = (
    study1
    .dropna(subset=['raw_current_area_of_residence'])
    .groupby(
        ['model', 'tone', 'raw_current_area_of_residence']
    )
    .size()
    .rename('Count')
    .reset_index()
    .sort_values(
        ['model', 'tone', 'Count'],
        ascending=[True, True, False]
    )
    .groupby(['model', 'tone'])
    .head(15)
)

top_hometowns.to_csv(
    TABLE_DIR / 'Top_Hometowns.csv',
    index=False
)
top_residences.to_csv(
    TABLE_DIR / 'Top_Residences.csv',
    index=False
)
# ================================================================
# 24. NAME ANALYSIS
# ================================================================
name_summary = (
    study1
    .dropna(
        subset=['raw_name']
    )
    .groupby(
        ['model', 'tone']
    )
    .agg(
        Parsed_Names=('raw_name', 'count'),
        Unique_Names=('raw_name', 'nunique')
    )
    .reset_index()
)
top_names = (
    study1
    .dropna(
        subset=['raw_name']
    )
    .groupby(
        [
            'model',
            'tone',
            'raw_name'
        ]
    )
    .size()
    .rename('Count')
    .reset_index()
    .sort_values(
        ['model', 'tone', 'Count'],
        ascending=[True, True, False]
    )
    .groupby(
        ['model', 'tone']
    )
    .head(15)
)
name_summary.to_csv(
    TABLE_DIR / 'Name_Summary.csv',
    index=False
)
top_names.to_csv(
    TABLE_DIR / 'Most_Common_Names.csv',
    index=False
)
# ================================================================
# 25. GENDER-PRESENTATION BREAKDOWN
# ================================================================
print("\n[13/16] Gender-presentation stratified analysis...")
gender_income_rows = []
for model in MODEL_ORDER:
    for gender in GENDER_ORDER:
        temp = (
            income_identity[
                income_identity['model'].eq(model)
                &
                income_identity['gender'].eq(gender)
            ]
            .pivot(
                index='identity',
                columns='tone',
                values='income_bdt'
            )
        )
        result = paired_statistics(
            temp.get('Light'),
            temp.get('Dark')
        )
        gender_income_rows.append({
            'model': model,
            'gender': gender,
            **result
        })
gender_income = pd.DataFrame(
    gender_income_rows
)
gender_education_rows = []
for model in MODEL_ORDER:
    for gender in GENDER_ORDER:
        temp = (
            education_identity[
                education_identity['model'].eq(model)
                &
                education_identity['gender'].eq(gender)
            ]
            .pivot(
                index='identity',
                columns='tone',
                values='bachelor_plus'
            )
        )
        result = paired_statistics(
            temp.get('Light'),
            temp.get('Dark')
        )
        gender_education_rows.append({
            'model': model,
            'gender': gender,
            **result
        })
gender_education = pd.DataFrame(
    gender_education_rows
)
gender_income.to_csv(
    TABLE_DIR / 'Income_By_Gender_Presentation.csv',
    index=False
)
gender_education.to_csv(
    TABLE_DIR / 'Education_By_Gender_Presentation.csv',
    index=False
)
# ------------------------------------------------
# FIGURE: GENDER-PRESENTATION INCOME GAP
# ------------------------------------------------
x = np.arange(
    len(MODEL_ORDER)
)
width = 0.36
fig, ax = plt.subplots(
    figsize=(9, 5.5)
)
for offset, gender in zip(
    [-width / 2, width / 2],
    GENDER_ORDER
):
    d = (
        gender_income[
            gender_income['gender'].eq(gender)
        ]
        .set_index('model')
        .loc[MODEL_ORDER]
        .reset_index()
    )
    means = d[
        'mean_difference_light_minus_dark'
    ].to_numpy()
    low = d[
        'bootstrap_95ci_low'
    ].to_numpy()
    high = d[
        'bootstrap_95ci_high'
    ].to_numpy()
    ax.bar(
        x + offset,
        means,
        width,
        label=gender,
        yerr=np.vstack([
            means - low,
            high - means
        ]),
        capsize=4
    )
ax.axhline(
    0,
    linewidth=1
)
ax.set_xticks(x)
ax.set_xticklabels(
    MODEL_ORDER
)
ax.set_ylabel(
    'Light − Dark income gap (BDT)'
)
ax.set_title(
    'Study 1: Income Skin-Tone Gap by Gender Presentation'
)
ax.legend(
    title='Pair group'
)
ax.grid(
    axis='y',
    alpha=0.25
)
fig.tight_layout()
fig.savefig(
    FIG_DIR / 'Income_Gap_By_Gender.png',
    bbox_inches='tight'
)
fig.savefig(
    FIG_DIR / 'Income_Gap_By_Gender.pdf',
    bbox_inches='tight'
)
plt.show()
# ================================================================
# 26. GENDER × SKIN-TONE INTERACTION ROBUSTNESS
# ================================================================
print("\n[14/16] Testing skin-tone × gender-presentation interaction...")

def binary_gee_is_sparse(data, group_cols):
    # Practical safeguard against complete/quasi-complete separation.
    summary = (
        data
        .groupby(group_cols)['bachelor_plus']
        .agg(['sum', 'count'])
        .reset_index()
    )
    summary['zeros'] = summary['count'] - summary['sum']
    summary['minority_count'] = summary[['sum', 'zeros']].min(axis=1)
    return bool((summary['minority_count'] < 2).any())

gender_interaction_results = []
try:
    import statsmodels.formula.api as smf
    from statsmodels.genmod.cov_struct import Exchangeable
    from statsmodels.genmod.families import Gaussian, Binomial

    interaction_data = study1.copy()
    interaction_data['tone_light'] = (
        interaction_data['tone'].eq('Light').astype(int)
    )
    interaction_data['gender_female'] = (
        interaction_data['gender'].eq('Female').astype(int)
    )
    interaction_term = 'tone_light:gender_female'

    for model in MODEL_ORDER:
        d = interaction_data[
            interaction_data['model'].eq(model)
        ].copy()

        # Monthly income interaction.
        try:
            fit = smf.gee(
                'income_bdt ~ tone_light * gender_female + C(trial)',
                groups='identity',
                data=d.dropna(subset=['income_bdt']),
                cov_struct=Exchangeable(),
                family=Gaussian()
            ).fit()
            ci = fit.conf_int().loc[interaction_term]
            gender_interaction_results.append({
                'Model': model,
                'Outcome': 'Monthly Income',
                'Interaction Term':
                    'Tone Light × Female Presentation',
                'Interaction Coefficient':
                    fit.params[interaction_term],
                '95% CI Low': ci.iloc[0],
                '95% CI High': ci.iloc[1],
                'P Value': fit.pvalues[interaction_term],
                'Interpretation Scale':
                    'BDT difference-in-differences',
                'Status': 'OK'
            })
        except Exception as e:
            gender_interaction_results.append({
                'Model': model,
                'Outcome': 'Monthly Income',
                'Interaction Term':
                    'Tone Light × Female Presentation',
                'Interaction Coefficient': np.nan,
                '95% CI Low': np.nan,
                '95% CI High': np.nan,
                'P Value': np.nan,
                'Interpretation Scale':
                    'BDT difference-in-differences',
                'Status': f'Fit failed: {type(e).__name__}'
            })

        # Bachelor+ interaction.
        if binary_gee_is_sparse(d, ['tone', 'gender']):
            gender_interaction_results.append({
                'Model': model,
                'Outcome': 'Bachelor-or-Higher',
                'Interaction Term':
                    'Tone Light × Female Presentation',
                'Interaction Coefficient': np.nan,
                '95% CI Low': np.nan,
                '95% CI High': np.nan,
                'P Value': np.nan,
                'Interpretation Scale': 'log-odds interaction',
                'Status':
                    'Skipped: sparse outcome / quasi-complete separation'
            })
        else:
            try:
                fit = smf.gee(
                    'bachelor_plus ~ tone_light * gender_female + C(trial)',
                    groups='identity',
                    data=d,
                    cov_struct=Exchangeable(),
                    family=Binomial()
                ).fit()
                ci = fit.conf_int().loc[interaction_term]
                gender_interaction_results.append({
                    'Model': model,
                    'Outcome': 'Bachelor-or-Higher',
                    'Interaction Term':
                        'Tone Light × Female Presentation',
                    'Interaction Coefficient':
                        fit.params[interaction_term],
                    '95% CI Low': ci.iloc[0],
                    '95% CI High': ci.iloc[1],
                    'P Value': fit.pvalues[interaction_term],
                    'Interpretation Scale':
                        'log-odds interaction',
                    'Status': 'OK'
                })
            except Exception as e:
                gender_interaction_results.append({
                    'Model': model,
                    'Outcome': 'Bachelor-or-Higher',
                    'Interaction Term':
                        'Tone Light × Female Presentation',
                    'Interaction Coefficient': np.nan,
                    '95% CI Low': np.nan,
                    '95% CI High': np.nan,
                    'P Value': np.nan,
                    'Interpretation Scale':
                        'log-odds interaction',
                    'Status': f'Fit failed: {type(e).__name__}'
                })
except Exception as e:
    print("Gender interaction section skipped:", e)

gender_interaction_results = pd.DataFrame(
    gender_interaction_results
)
gender_interaction_results.to_csv(
    STAT_DIR / 'Study1_Gender_Interaction.csv',
    index=False
)

# ================================================================
# 27. GEE ROBUSTNESS CHECKS
# ================================================================
print("\n[15/16] Running repeated-measures robustness checks (GEE)...")
gee_results = []
try:
    import statsmodels.formula.api as smf
    from statsmodels.genmod.cov_struct import Exchangeable
    from statsmodels.genmod.families import Gaussian, Binomial

    gee_data = study1.copy()
    gee_data['tone_light'] = gee_data['tone'].eq('Light').astype(int)
    gee_data['gender_female'] = gee_data['gender'].eq('Female').astype(int)

    for model in MODEL_ORDER:
        d = gee_data[gee_data['model'].eq(model)].copy()

        # Income.
        try:
            fit = smf.gee(
                'income_bdt ~ tone_light + gender_female + C(trial)',
                groups='identity',
                data=d.dropna(subset=['income_bdt']),
                cov_struct=Exchangeable(),
                family=Gaussian()
            ).fit()
            ci = fit.conf_int().loc['tone_light']
            gee_results.append({
                'Model': model,
                'Outcome': 'Monthly Income',
                'Tone Light Coefficient': fit.params['tone_light'],
                '95% CI Low': ci.iloc[0],
                '95% CI High': ci.iloc[1],
                'P Value': fit.pvalues['tone_light'],
                'Status': 'OK'
            })
        except Exception as e:
            gee_results.append({
                'Model': model,
                'Outcome': 'Monthly Income',
                'Tone Light Coefficient': np.nan,
                '95% CI Low': np.nan,
                '95% CI High': np.nan,
                'P Value': np.nan,
                'Status': f'Fit failed: {type(e).__name__}'
            })

        # Bachelor+.
        if binary_gee_is_sparse(d, ['tone']):
            gee_results.append({
                'Model': model,
                'Outcome': 'Bachelor-or-Higher',
                'Tone Light Coefficient': np.nan,
                '95% CI Low': np.nan,
                '95% CI High': np.nan,
                'P Value': np.nan,
                'Status':
                    'Skipped: sparse outcome / quasi-complete separation'
            })
        else:
            try:
                fit = smf.gee(
                    'bachelor_plus ~ tone_light + gender_female + C(trial)',
                    groups='identity',
                    data=d,
                    cov_struct=Exchangeable(),
                    family=Binomial()
                ).fit()
                ci = fit.conf_int().loc['tone_light']
                gee_results.append({
                    'Model': model,
                    'Outcome': 'Bachelor-or-Higher',
                    'Tone Light Coefficient':
                        fit.params['tone_light'],
                    '95% CI Low': ci.iloc[0],
                    '95% CI High': ci.iloc[1],
                    'P Value': fit.pvalues['tone_light'],
                    'Status': 'OK'
                })
            except Exception as e:
                gee_results.append({
                    'Model': model,
                    'Outcome': 'Bachelor-or-Higher',
                    'Tone Light Coefficient': np.nan,
                    '95% CI Low': np.nan,
                    '95% CI High': np.nan,
                    'P Value': np.nan,
                    'Status': f'Fit failed: {type(e).__name__}'
                })
except Exception as e:
    print("GEE section skipped:", e)

gee_results = pd.DataFrame(gee_results)
gee_results.to_csv(
    STAT_DIR / 'Study1_GEE_Robustness.csv',
    index=False
)
# ================================================================
# 28. CONSOLIDATED RESULTS WORKBOOK
# ================================================================
print("\n[16/16] Creating consolidated Study 1 workbook...")
results_file = (
    TABLE_DIR /
    'Study1_Results.xlsx'
)
with pd.ExcelWriter(
    results_file,
    engine='openpyxl'
) as writer:
    validation.to_excel(
        writer,
        sheet_name='Validation',
        index=False
    )
    parsing_completeness.to_excel(
        writer,
        sheet_name='Parsing',
        index=False
    )
    income_descriptive.to_excel(
        writer,
        sheet_name='Income Descriptive',
        index=False
    )
    income_paired.to_excel(
        writer,
        sheet_name='Income Paired',
        index=False
    )
    education_descriptive.to_excel(
        writer,
        sheet_name='Education Bachelor+',
        index=False
    )
    education_paired.to_excel(
        writer,
        sheet_name='Education Paired',
        index=False
    )
    education_distribution.to_excel(
        writer,
        sheet_name='Education Distribution',
        index=False
    )
    llama_dark_bachelor_audit.to_excel(
        writer,
        sheet_name='Llama Edu Audit',
        index=False
    )
    age_descriptive.to_excel(
        writer,
        sheet_name='Age Descriptive',
        index=False
    )
    age_paired.to_excel(
        writer,
        sheet_name='Age Paired',
        index=False
    )
    occupation_distribution.to_excel(
        writer,
        sheet_name='Occupation',
        index=False
    )
    top_occupations.to_excel(
        writer,
        sheet_name='Top Occupations',
        index=False
    )
    experience_presence_summary.to_excel(
        writer,
        sheet_name='Experience Presence',
        index=False
    )
    hometown_dhaka_summary.to_excel(
        writer,
        sheet_name='Hometown Dhaka',
        index=False
    )
    top_hometowns.to_excel(
        writer,
        sheet_name='Top Hometowns',
        index=False
    )
    top_residences.to_excel(
        writer,
        sheet_name='Top Residences',
        index=False
    )
    name_summary.to_excel(
        writer,
        sheet_name='Name Summary',
        index=False
    )
    top_names.to_excel(
        writer,
        sheet_name='Top Names',
        index=False
    )
    gender_income.to_excel(
        writer,
        sheet_name='Income Gender',
        index=False
    )
    gender_education.to_excel(
        writer,
        sheet_name='Education Gender',
        index=False
    )
    gender_interaction_results.to_excel(
        writer,
        sheet_name='Gender Interaction',
        index=False
    )
    gee_results.to_excel(
        writer,
        sheet_name='GEE Robustness',
        index=False
    )
# ================================================================
# 29. METHOD NOTES
# ================================================================
method_notes = f"""
FYDP STUDY 1 ANALYSIS NOTES
===========================
Source:
{DATA_FILE}

Primary models:
Gemini
GPT
DeepSeek
Llama

Study design:
100 images
50 matched identities
2 skin-tone versions per identity
3 trials per image/model
4 primary models
Total primary responses = 1,200

Main inferential unit:
The three trials are averaged within each identity and skin tone.
Therefore the main paired comparisons use 50 matched identities,
rather than treating 300 responses per model as independent people.

Primary outcomes:
1. Monthly Income
2. Bachelor-or-Higher Education

Primary statistics:
Bootstrap 95% confidence intervals
Paired Wilcoxon signed-rank test
Paired Cohen's dz

Repeated-measures robustness:
GEE is used for income and Bachelor-or-Higher outcomes.
Binary GEE estimates are automatically skipped when outcome cells are
too sparse and complete/quasi-complete separation would make the
coefficient unstable. This especially protects the near-ceiling Gemini
Bachelor-or-Higher result from being misinterpreted.

Gender-presentation analysis:
Matched tone effects are reported separately for the Male- and
Female-presentation identity groups.
A secondary GEE skin-tone × gender-presentation interaction is also
estimated to test whether the tone effect differs between those groups.
This is not a direct male-vs-female hiring or social comparison.

Secondary descriptive outcomes:
Age
Occupation category
Previous work-experience presence
Hometown
Current residence
Name generation

Occupation:
Occupation categories are exploratory keyword-based sector groupings.
No occupation-prestige score is created.
Top raw occupation strings are exported for auditability.

Work experience:
Only presence/absence is retained for reporting.
Duration-in-years analysis is excluded because durations were not
consistently recoverable across response formats.

Residence:
Current residence is NOT reduced to a simple Dhaka/non-Dhaka flag.
Many valid Dhaka neighborhoods (e.g., Mirpur, Dhanmondi, Uttara,
Banani) may be written without the word "Dhaka".
Raw top-residence distributions are therefore retained instead.
No neighborhood-prestige score is created.

Education audit:
The final source-derived count for Llama dark-skin Bachelor-or-Higher
responses is 12/150. The matching response-level cases are exported to
Llama_Dark_BachelorPlus_Audit.csv.

Qwen is excluded from the defined four-model primary comparison.
"""
with open(
    STAT_DIR / 'Study1_Method_Notes.txt',
    'w',
    encoding='utf-8'
) as f:
    f.write(
        method_notes
    )
# ================================================================
# 30. FINAL SUMMARY
# ================================================================
print("\n" + "=" * 70)
print("STUDY 1 ANALYSIS COMPLETE")
print("=" * 70)
print("\nVALIDATION:")
display(validation)
print("\nPARSING COMPLETENESS:")
display(parsing_completeness)
print("\nPRIMARY INCOME RESULTS:")
display(
    income_paired[
        [
            'model',
            'mean_light',
            'mean_dark',
            'mean_difference_light_minus_dark',
            'percent_difference_vs_dark',
            'light_higher_pairs',
            'dark_higher_pairs',
            'ties',
            'bootstrap_95ci_low',
            'bootstrap_95ci_high',
            'wilcoxon_p',
            'cohen_dz'
        ]
    ]
)
print("\nPRIMARY EDUCATION RESULTS:")
display(
    education_descriptive[
        [
            'model',
            'tone',
            'Bachelor_Plus_Count',
            'N',
            'Bachelor_Plus_Percent'
        ]
    ]
)
print("\nGENDER × SKIN-TONE INTERACTION ROBUSTNESS:")
display(gender_interaction_results)
print("\nGEE ROBUSTNESS:")
display(gee_results)
print("\nOUTPUT LOCATIONS:")
print("Cleaned data :", CLEAN_DIR)
print("Figures      :", FIG_DIR)
print("Statistics   :", STAT_DIR)
print("Tables       :", TABLE_DIR)
print("Results XLSX :", results_file)
print("\nFIGURES CREATED:")
for f in sorted(FIG_DIR.glob('*')):
    print(" •", f.name)
print("\nTABLE/RESULT FILES CREATED:")
for f in sorted(TABLE_DIR.glob('*')):
    print(" •", f.name)
print("\nSTATISTICS FILES CREATED:")
for f in sorted(STAT_DIR.glob('*')):
    print(" •", f.name)
print("\n✓ Finished.")
```
